```python
# ============================================================
# FESR-ViT FOR POTATO DATASET
# ============================================================
#
# FESR = Feature Energy-Sensitivity Redistribution
#
# Pipeline:
#
#                 POTATO ZIP
#                     |
#          -----------------------
#          |          |          |
#         RGB     Pseudo-Depth  Synthetic
#          |          |          |
#          -------- Feature Extraction
#                     |
#               Feature Fusion
#                     |
#              Energy Estimation
#                     |
#              Classification
#                     |
#                  Loss L
#                     |
#             Task Sensitivity
#              dL / dF_j
#                     |
#       Energy-Sensitivity Integration
#                     |
#           Adaptive Redistribution
#                     |
#          Balanced Representation
#                     |
#                    ViT
#                     |
#                Classifier
#                     |
#              Final Prediction
#
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import os
import zipfile
import random
import warnings

import numpy as np
import matplotlib.pyplot as plt

from PIL import Image, ImageOps, ImageEnhance

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import (
    Dataset,
    DataLoader,
    WeightedRandomSampler
)

from torchvision import transforms

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

from tqdm.auto import tqdm


warnings.filterwarnings("ignore")


# ============================================================
# 2. CONFIGURATION
# ============================================================

ZIP_PATH = "./Potato.zip"

EXTRACT_PATH = "./Potato_extracted"

IMAGE_SIZE = 224

BATCH_SIZE = 16

NUM_EPOCHS = 20

LEARNING_RATE = 1e-4

WEIGHT_DECAY = 1e-4

NUM_CLASSES = 3

SEED = 42


# ============================================================
# FESR PARAMETERS
# ============================================================

ALPHA = 0.30

EPSILON = 1e-8

LAMBDA_REDISTRIBUTION = 0.10

LAMBDA_CONSISTENCY = 0.05

LAMBDA_ENTROPY = 0.01


# ============================================================
# 3. CLASS NAMES
# ============================================================

CLASS_NAMES = [

    "Early Blight",

    "Late Blight",

    "Healthy"

]


# ============================================================
# 4. REPRODUCIBILITY
# ============================================================

def set_seed(seed):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():

        torch.cuda.manual_seed(seed)

        torch.cuda.manual_seed_all(seed)


set_seed(SEED)


# ============================================================
# 5. GPU CONFIGURATION
# ============================================================

if torch.cuda.is_available():

    DEVICE = torch.device("cuda")

    print("=" * 70)

    print("GPU DETECTED")

    print("=" * 70)

    print(
        "GPU:",
        torch.cuda.get_device_name(0)
    )

    print(
        "CUDA:",
        torch.version.cuda
    )

    print(
        "GPU Memory:",
        round(
            torch.cuda.get_device_properties(0).total_memory
            / (1024 ** 3),
            2
        ),
        "GB"
    )

    # Fast convolution algorithms
    torch.backends.cudnn.benchmark = True

    # TF32 for supported NVIDIA GPUs
    torch.backends.cuda.matmul.allow_tf32 = True

    torch.backends.cudnn.allow_tf32 = True

else:

    DEVICE = torch.device("cpu")

    print("=" * 70)

    print("GPU NOT DETECTED")

    print("Running on CPU")

    print("=" * 70)


print(
    "\nSelected device:",
    DEVICE
)


# ============================================================
# 6. MIXED PRECISION
# ============================================================

USE_AMP = (

    DEVICE.type == "cuda"

)


if USE_AMP:

    scaler = torch.amp.GradScaler(
        "cuda"
    )

else:

    scaler = None


# ============================================================
# 7. CHECK ZIP
# ============================================================

if not os.path.exists(ZIP_PATH):

    raise FileNotFoundError(

        f"\nCould not find:\n{ZIP_PATH}\n\n"

        "Place Potato.zip in the same directory "
        "as this notebook/script."

    )


# ============================================================
# 8. EXTRACT DATASET
# ============================================================

if not os.path.exists(EXTRACT_PATH):

    print(
        "\nExtracting Potato.zip..."
    )

    with zipfile.ZipFile(

        ZIP_PATH,

        "r"

    ) as zip_ref:

        zip_ref.extractall(
            EXTRACT_PATH
        )

    print(
        "Extraction completed."
    )

else:

    print(
        "\nDataset already extracted."
    )


# ============================================================
# 9. FIND CLASS FOLDERS
# ============================================================

def normalize_name(name):

    return (

        name

        .lower()

        .replace(
            "_",
            " "
        )

        .replace(
            "-",
            " "
        )

    )


def get_class_label(folder_name):

    name = normalize_name(
        folder_name
    )


    if (

        "early" in name

        and

        "blight" in name

    ):

        return 0


    if (

        "late" in name

        and

        "blight" in name

    ):

        return 1


    if "healthy" in name:

        return 2


    return None


# ============================================================
# 10. COLLECT IMAGE PATHS
# ============================================================

def collect_images(root):

    image_extensions = (

        ".jpg",

        ".jpeg",

        ".png",

        ".JPG",

        ".JPEG",

        ".PNG"

    )


    paths = []

    labels = []


    for current_root, dirs, files in os.walk(root):

        folder = os.path.basename(
            current_root
        )


        label = get_class_label(
            folder
        )


        if label is None:

            continue


        for file in files:

            if file.endswith(
                image_extensions
            ):

                paths.append(

                    os.path.join(

                        current_root,

                        file

                    )

                )

                labels.append(
                    label
                )


    return paths, labels


image_paths, labels = collect_images(
    EXTRACT_PATH
)


# ============================================================
# 11. DATASET CHECK
# ============================================================

if len(image_paths) == 0:

    raise RuntimeError(

        "\nNo Potato images were found.\n\n"

        "Expected class folders containing names such as:\n"

        "Potato___Early_blight\n"

        "Potato___Late_blight\n"

        "Potato___healthy\n\n"

        f"Extracted directory:\n{EXTRACT_PATH}"

    )


print("\n")
print("=" * 70)
print("DATASET")
print("=" * 70)

print(
    "Total images:",
    len(image_paths)
)


for i, class_name in enumerate(
    CLASS_NAMES
):

    count = labels.count(i)

    print(
        f"{class_name}: {count}"
    )


# ============================================================
# 12. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

train_paths, temp_paths, train_labels, temp_labels = (

    train_test_split(

        image_paths,

        labels,

        test_size=0.30,

        stratify=labels,

        random_state=SEED

    )

)


val_paths, test_paths, val_labels, test_labels = (

    train_test_split(

        temp_paths,

        temp_labels,

        test_size=0.50,

        stratify=temp_labels,

        random_state=SEED

    )

)


print("\n")
print("=" * 70)
print("DATA SPLIT")
print("=" * 70)

print(
    "Training:",
    len(train_paths)
)

print(
    "Validation:",
    len(val_paths)
)

print(
    "Testing:",
    len(test_paths)
)


# ============================================================
# 13. TRANSFORMS
# ============================================================

train_transform = transforms.Compose([

    transforms.Resize(

        (IMAGE_SIZE, IMAGE_SIZE)

    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomVerticalFlip(
        p=0.2
    ),

    transforms.RandomRotation(
        15
    ),

    transforms.ColorJitter(

        brightness=0.20,

        contrast=0.20,

        saturation=0.20,

        hue=0.05

    ),

    transforms.ToTensor(),

    transforms.Normalize(

        mean=[

            0.485,

            0.456,

            0.406

        ],

        std=[

            0.229,

            0.224,

            0.225

        ]

    )

])


test_transform = transforms.Compose([

    transforms.Resize(

        (IMAGE_SIZE, IMAGE_SIZE)

    ),

    transforms.ToTensor(),

    transforms.Normalize(

        mean=[

            0.485,

            0.456,

            0.406

        ],

        std=[

            0.229,

            0.224,

            0.225

        ]

    )

])


# ============================================================
# 14. PSEUDO-DEPTH GENERATION
# ============================================================

def create_pseudo_depth(image):

    """

    Creates a structural pseudo-depth representation.

    NOTE:
    This is NOT real depth sensor data.

    It is used to create a second structural modality
    for experimentation.

    """


    gray = ImageOps.grayscale(
        image
    )


    gray = gray.resize(

        (
            IMAGE_SIZE,

            IMAGE_SIZE

        )

    )


    gray = ImageEnhance.Contrast(

        gray

    ).enhance(

        1.5

    )


    depth = np.asarray(

        gray,

        dtype=np.float32

    )


    depth = (

        depth

        /

        255.0

    )


    depth = torch.from_numpy(
        depth
    )


    depth = depth.unsqueeze(
        0
    )


    return depth


# ============================================================
# 15. SYNTHETIC IMAGE GENERATION
# ============================================================

def create_synthetic_image(image):

    """

    Synthetic augmentation.

    NOTE:
    This is NOT a trained GAN.

    A real GAN generator can replace this function later.

    """


    img = image.copy()


    choice = random.randint(
        0,
        4
    )


    if choice == 0:

        img = ImageEnhance.Brightness(
            img
        ).enhance(

            random.uniform(
                0.75,
                1.25
            )

        )


    elif choice == 1:

        img = ImageEnhance.Contrast(
            img
        ).enhance(

            random.uniform(
                0.75,
                1.25
            )

        )


    elif choice == 2:

        img = ImageEnhance.Color(
            img
        ).enhance(

            random.uniform(
                0.70,
                1.30
            )

        )


    elif choice == 3:

        img = ImageOps.mirror(
            img
        )


    else:

        img = img.rotate(

            random.uniform(
                -15,
                15
            )

        )


    return img


# ============================================================
# 16. DATASET CLASS
# ============================================================

class PotatoDataset(Dataset):

    def __init__(

        self,

        paths,

        labels,

        transform

    ):

        self.paths = paths

        self.labels = labels

        self.transform = transform


    def __len__(self):

        return len(
            self.paths
        )


    def __getitem__(

        self,

        index

    ):

        path = self.paths[
            index
        ]

        label = self.labels[
            index
        ]


        image = Image.open(
            path
        ).convert(
            "RGB"
        )


        original = image.copy()


        # --------------------------------------------
        # RGB
        # --------------------------------------------

        rgb = self.transform(
            image
        )


        # --------------------------------------------
        # PSEUDO DEPTH
        # --------------------------------------------

        depth = create_pseudo_depth(
            original
        )


        # --------------------------------------------
        # SYNTHETIC
        # --------------------------------------------

        synthetic_image = (

            create_synthetic_image(
                original
            )

        )


        synthetic = self.transform(
            synthetic_image
        )


        return (

            rgb,

            depth,

            synthetic,

            torch.tensor(

                label,

                dtype=torch.long

            )

        )


# ============================================================
# 17. DATASETS
# ============================================================

train_dataset = PotatoDataset(

    train_paths,

    train_labels,

    train_transform

)


val_dataset = PotatoDataset(

    val_paths,

    val_labels,

    test_transform

)


test_dataset = PotatoDataset(

    test_paths,

    test_labels,

    test_transform

)


# ============================================================
# 18. CLASS BALANCING
# ============================================================

class_counts = np.bincount(

    train_labels,

    minlength=NUM_CLASSES

)


class_weights = (

    len(train_labels)

    /

    (

        NUM_CLASSES

        *

        class_counts

    )

)


sample_weights = [

    class_weights[label]

    for label in train_labels

]


sampler = WeightedRandomSampler(

    weights=torch.tensor(

        sample_weights,

        dtype=torch.double

    ),

    num_samples=len(
        sample_weights
    ),

    replacement=True

)


# ============================================================
# 19. DATALOADERS
# ============================================================

# Google Colab/Linux:
NUM_WORKERS = 2

# If Windows multiprocessing causes problems,
# set NUM_WORKERS = 0.


train_loader = DataLoader(

    train_dataset,

    batch_size=BATCH_SIZE,

    sampler=sampler,

    num_workers=NUM_WORKERS,

    pin_memory=(DEVICE.type == "cuda"),

    persistent_workers=(
        NUM_WORKERS > 0
    )

)


val_loader = DataLoader(

    val_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=(DEVICE.type == "cuda"),

    persistent_workers=(
        NUM_WORKERS > 0
    )

)


test_loader = DataLoader(

    test_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=(DEVICE.type == "cuda"),

    persistent_workers=(
        NUM_WORKERS > 0
    )

)


# ============================================================
# 20. FEATURE ENCODER
# ============================================================

class FeatureEncoder(nn.Module):

    def __init__(

        self,

        input_channels,

        embed_dim=192

    ):

        super().__init__()


        self.encoder = nn.Sequential(

            nn.Conv2d(

                input_channels,

                64,

                kernel_size=3,

                stride=2,

                padding=1

            ),

            nn.BatchNorm2d(
                64
            ),

            nn.GELU(),


            nn.Conv2d(

                64,

                96,

                kernel_size=3,

                stride=2,

                padding=1

            ),

            nn.BatchNorm2d(
                96
            ),

            nn.GELU(),


            nn.Conv2d(

                96,

                128,

                kernel_size=3,

                stride=2,

                padding=1

            ),

            nn.BatchNorm2d(
                128
            ),

            nn.GELU(),


            nn.Conv2d(

                128,

                embed_dim,

                kernel_size=3,

                stride=2,

                padding=1

            ),

            nn.BatchNorm2d(
                embed_dim
            ),

            nn.GELU()

        )


    def forward(self, x):

        x = self.encoder(x)


        # B,C,H,W
        #
        # →
        #
        # B,N,C

        x = x.flatten(
            2
        )


        x = x.transpose(
            1,
            2
        )


        return x


# ============================================================
# 21. FESR-ViT
# ============================================================

class FESRViT(nn.Module):

    def __init__(

        self,

        num_classes=3,

        embed_dim=192,

        depth=6,

        heads=6

    ):

        super().__init__()


        self.embed_dim = embed_dim


        # --------------------------------------------
        # RGB branch
        # --------------------------------------------

        self.rgb_encoder = FeatureEncoder(

            3,

            embed_dim

        )


        # --------------------------------------------
        # Depth branch
        # --------------------------------------------

        self.depth_encoder = FeatureEncoder(

            1,

            embed_dim

        )


        # --------------------------------------------
        # Synthetic branch
        # --------------------------------------------

        self.synthetic_encoder = FeatureEncoder(

            3,

            embed_dim

        )


        # --------------------------------------------
        # Fusion
        # --------------------------------------------

        self.fusion = nn.Linear(

            embed_dim * 3,

            embed_dim

        )


        # --------------------------------------------
        # CLS token
        # --------------------------------------------

        self.cls_token = nn.Parameter(

            torch.zeros(

                1,

                1,

                embed_dim

            )

        )


        # Feature maps are 14 x 14 = 196 tokens
        # plus CLS token = 197

        self.position_embedding = nn.Parameter(

            torch.zeros(

                1,

                197,

                embed_dim

            )

        )


        # --------------------------------------------
        # Transformer
        # --------------------------------------------

        encoder_layer = nn.TransformerEncoderLayer(

            d_model=embed_dim,

            nhead=heads,

            dim_feedforward=embed_dim * 4,

            dropout=0.1,

            activation="gelu",

            batch_first=True,

            norm_first=True

        )


        self.transformer = nn.TransformerEncoder(

            encoder_layer,

            num_layers=depth

        )


        self.norm = nn.LayerNorm(

            embed_dim

        )


        # --------------------------------------------
        # Classifier
        # --------------------------------------------

        self.classifier = nn.Linear(

            embed_dim,

            num_classes

        )


        nn.init.trunc_normal_(

            self.cls_token,

            std=0.02

        )


        nn.init.trunc_normal_(

            self.position_embedding,

            std=0.02

        )


    # ========================================================
    # ENERGY
    # ========================================================

    def calculate_energy(

        self,

        features

    ):

        # F ∈ R^(B × N × D)

        energy = torch.mean(

            features ** 2,

            dim=1

        )


        return energy


    # ========================================================
    # ENERGY DISTRIBUTION
    # ========================================================

    def calculate_energy_distribution(

        self,

        features

    ):

        energy = self.calculate_energy(

            features

        )


        distribution = (

            energy

            /

            (

                energy.sum(

                    dim=1,

                    keepdim=True

                )

                + EPSILON

            )

        )


        return distribution


    # ========================================================
    # ENERGY-SENSITIVITY INTEGRATION
    # ========================================================

    def energy_sensitivity_integration(

        self,

        features,

        sensitivity

    ):

        energy = (

            self.calculate_energy_distribution(

                features

            )

        )


        sensitivity = (

            sensitivity

            /

            (

                sensitivity.sum(

                    dim=1,

                    keepdim=True

                )

                + EPSILON

            )

        )


        # ------------------------------------------------
        # Proposed integration
        #
        # T_j =
        #
        # (1-alpha) E_j
        # +
        # alpha S_j
        # ------------------------------------------------

        target = (

            (1.0 - ALPHA)

            *

            energy

            +

            ALPHA

            *

            sensitivity

        )


        target = target / (

            target.sum(

                dim=1,

                keepdim=True

            )

            + EPSILON

        )


        return (

            energy,

            sensitivity,

            target

        )


    # ========================================================
    # ADAPTIVE REDISTRIBUTION
    # ========================================================

    def adaptive_redistribution(

        self,

        features,

        sensitivity

    ):

        (

            energy,

            sensitivity_dist,

            target

        ) = self.energy_sensitivity_integration(

            features,

            sensitivity

        )


        # ------------------------------------------------
        # Mathematical redistribution
        #
        # w_j =
        #
        # sqrt(
        # T_j / E_j
        # )
        # ------------------------------------------------

        scale = torch.sqrt(

            (

                target

                +

                EPSILON

            )

            /

            (

                energy

                +

                EPSILON

            )

        )


        # Prevent numerical instability

        scale = torch.clamp(

            scale,

            min=0.25,

            max=4.0

        )


        balanced_features = (

            features

            *

            scale.unsqueeze(1)

        )


        return (

            balanced_features,

            energy,

            sensitivity_dist,

            target

        )


    # ========================================================
    # FORWARD
    # ========================================================

    def forward(

        self,

        rgb,

        depth,

        synthetic,

        sensitivity=None

    ):

        # ------------------------------------------------
        # Feature extraction
        # ------------------------------------------------

        F_R = self.rgb_encoder(
            rgb
        )


        F_D = self.depth_encoder(
            depth
        )


        F_S = self.synthetic_encoder(
            synthetic
        )


        # ------------------------------------------------
        # Token alignment
        # ------------------------------------------------

        n = min(

            F_R.shape[1],

            F_D.shape[1],

            F_S.shape[1]

        )


        F_R = F_R[
            :, :n, :
        ]


        F_D = F_D[
            :, :n, :
        ]


        F_S = F_S[
            :, :n, :
        ]


        # ------------------------------------------------
        # Feature fusion
        # ------------------------------------------------

        fused_features = torch.cat(

            [

                F_R,

                F_D,

                F_S

            ],

            dim=-1

        )


        fused_features = self.fusion(

            fused_features

        )


        # ------------------------------------------------
        # IMPORTANT FIX
        #
        # retain_grad() is called ONLY when:
        #
        # 1. Training
        # 2. Gradients enabled
        # 3. Tensor requires gradient
        # ------------------------------------------------

        if (

            self.training

            and

            torch.is_grad_enabled()

            and

            fused_features.requires_grad

        ):

            fused_features.retain_grad()


        # ------------------------------------------------
        # FESR redistribution
        # ------------------------------------------------

        if sensitivity is not None:

            (

                balanced_features,

                energy_distribution,

                sensitivity_distribution,

                target_distribution

            ) = self.adaptive_redistribution(

                fused_features,

                sensitivity

            )

        else:

            energy_distribution = (

                self.calculate_energy_distribution(

                    fused_features

                )

            )


            balanced_features = fused_features


            sensitivity_distribution = (
                energy_distribution
            )


            target_distribution = (
                energy_distribution
            )


        # ------------------------------------------------
        # CLS token
        # ------------------------------------------------

        B = balanced_features.shape[0]


        cls = self.cls_token.expand(

            B,

            -1,

            -1

        )


        tokens = torch.cat(

            [

                cls,

                balanced_features

            ],

            dim=1

        )


        # ------------------------------------------------
        # Position encoding
        # ------------------------------------------------

        tokens = (

            tokens

            +

            self.position_embedding[

                :, :tokens.shape[1], :

            ]

        )


        # ------------------------------------------------
        # ViT
        # ------------------------------------------------

        encoded = self.transformer(

            tokens

        )


        encoded = self.norm(

            encoded

        )


        # ------------------------------------------------
        # CLS representation
        # ------------------------------------------------

        representation = encoded[
            :,
            0
        ]


        # ------------------------------------------------
        # Classifier
        # ------------------------------------------------

        logits = self.classifier(

            representation

        )


        return {

            "logits":
                logits,

            "features":
                fused_features,

            "balanced_features":
                balanced_features,

            "energy_distribution":
                energy_distribution,

            "sensitivity_distribution":
                sensitivity_distribution,

            "target_distribution":
                target_distribution,

            "rgb_features":
                F_R,

            "depth_features":
                F_D,

            "synthetic_features":
                F_S,

            "representation":
                representation

        }


# ============================================================
# 22. CREATE MODEL
# ============================================================

model = FESRViT(

    num_classes=NUM_CLASSES,

    embed_dim=192,

    depth=6,

    heads=6

)


model = model.to(
    DEVICE
)


# ============================================================
# 23. MODEL INFORMATION
# ============================================================

total_params = sum(

    p.numel()

    for p in model.parameters()

)


trainable_params = sum(

    p.numel()

    for p in model.parameters()

    if p.requires_grad

)


print("\n")
print("=" * 70)
print("MODEL INFORMATION")
print("=" * 70)

print(
    "Total parameters:",
    f"{total_params:,}"
)

print(
    "Trainable parameters:",
    f"{trainable_params:,}"
)


# ============================================================
# 24. LOSSES
# ============================================================

classification_loss = nn.CrossEntropyLoss(

    label_smoothing=0.1

)


def kl_divergence(

    p,

    q

):

    p = torch.clamp(

        p,

        min=EPSILON

    )


    q = torch.clamp(

        q,

        min=EPSILON

    )


    return torch.mean(

        torch.sum(

            p

            *

            torch.log(

                p / q

            ),

            dim=1

        )

    )


def js_divergence(

    p,

    q

):

    p = torch.clamp(

        p,

        min=EPSILON

    )


    q = torch.clamp(

        q,

        min=EPSILON

    )


    m = (

        0.5

        *

        (

            p + q

        )

    )


    return (

        0.5

        *

        (

            kl_divergence(
                p,
                m
            )

            +

            kl_divergence(
                q,
                m
            )

        )

    )


def entropy_regularization(

    p

):

    p = torch.clamp(

        p,

        min=EPSILON

    )


    entropy = -torch.sum(

        p

        *

        torch.log(p),

        dim=1

    )


    return -torch.mean(
        entropy
    )


# ============================================================
# 25. OPTIMIZER
# ============================================================

optimizer = torch.optim.AdamW(

    model.parameters(),

    lr=LEARNING_RATE,

    weight_decay=WEIGHT_DECAY

)


scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(

    optimizer,

    T_max=NUM_EPOCHS

)


# ============================================================
# 26. TASK SENSITIVITY
# ============================================================

def calculate_task_sensitivity(

    features,

    loss

):

    """

    S_j = sqrt(
        mean(
            (dL/dF_j)^2
        )
    )

    """

    gradients = torch.autograd.grad(

        outputs=loss,

        inputs=features,

        retain_graph=True,

        create_graph=False,

        allow_unused=True

    )[0]


    if gradients is None:

        sensitivity = torch.ones(

            features.shape[0],

            features.shape[2],

            device=features.device

        )


    else:

        sensitivity = torch.sqrt(

            torch.mean(

                gradients.detach() ** 2,

                dim=1

            )

            +

            EPSILON

        )


    return sensitivity


# ============================================================
# 27. TRAINING FUNCTION
# ============================================================

def train_one_epoch():

    model.train()


    running_loss = 0.0

    running_correct = 0

    running_total = 0


    progress = tqdm(

        train_loader,

        desc="Training",

        leave=True

    )


    for (

        rgb,

        depth,

        synthetic,

        labels

    ) in progress:


        # ====================================================
        # GPU TRANSFER
        # ====================================================

        rgb = rgb.to(

            DEVICE,

            non_blocking=True

        )


        depth = depth.to(

            DEVICE,

            non_blocking=True

        )


        synthetic = synthetic.to(

            DEVICE,

            non_blocking=True

        )


        labels = labels.to(

            DEVICE,

            non_blocking=True

        )


        optimizer.zero_grad(

            set_to_none=True

        )


        # ====================================================
        # FIRST FORWARD PASS
        #
        # Used to determine task sensitivity.
        # ====================================================

        with torch.autocast(

            device_type=DEVICE.type,

            dtype=torch.float16,

            enabled=USE_AMP

        ):

            first_output = model(

                rgb,

                depth,

                synthetic,

                sensitivity=None

            )


            first_logits = first_output[
                "logits"
            ]


            first_loss = classification_loss(

                first_logits,

                labels

            )


        # ====================================================
        # TASK SENSITIVITY
        # ====================================================

        features = first_output[
            "features"
        ]


        if USE_AMP:

            scaled_loss = (

                scaler.scale(
                    first_loss
                )

            )


            gradients = torch.autograd.grad(

                scaled_loss,

                features,

                retain_graph=True,

                create_graph=False,

                allow_unused=True

            )[0]


            if gradients is not None:

                gradients = (

                    gradients

                    /

                    scaler.get_scale()

                )

        else:

            gradients = torch.autograd.grad(

                first_loss,

                features,

                retain_graph=True,

                create_graph=False,

                allow_unused=True

            )[0]


        if gradients is None:

            sensitivity = torch.ones(

                features.shape[0],

                features.shape[2],

                device=DEVICE

            )

        else:

            sensitivity = torch.sqrt(

                torch.mean(

                    gradients.detach() ** 2,

                    dim=1

                )

                +

                EPSILON

            )


        # ====================================================
        # SECOND FORWARD PASS
        #
        # FESR redistribution
        # ====================================================

        with torch.autocast(

            device_type=DEVICE.type,

            dtype=torch.float16,

            enabled=USE_AMP

        ):

            output = model(

                rgb,

                depth,

                synthetic,

                sensitivity=sensitivity

            )


            logits = output[
                "logits"
            ]


            # --------------------------------------------
            # Classification loss
            # --------------------------------------------

            loss_cls = classification_loss(

                logits,

                labels

            )


            # --------------------------------------------
            # Energy redistribution loss
            # --------------------------------------------

            energy = output[

                "energy_distribution"

            ]


            target = output[

                "target_distribution"

            ]


            loss_red = kl_divergence(

                target,

                energy

            )


            # --------------------------------------------
            # Synthetic consistency
            # --------------------------------------------

            synthetic_features = output[

                "synthetic_features"

            ]


            synthetic_energy = (

                model.calculate_energy_distribution(

                    synthetic_features

                )

            )


            loss_consistency = js_divergence(

                energy,

                synthetic_energy

            )


            # --------------------------------------------
            # Entropy
            # --------------------------------------------

            loss_entropy = entropy_regularization(

                target

            )


            # --------------------------------------------
            # TOTAL LOSS
            # --------------------------------------------

            loss = (

                loss_cls

                +

                LAMBDA_REDISTRIBUTION
                *
                loss_red

                +

                LAMBDA_CONSISTENCY
                *
                loss_consistency

                +

                LAMBDA_ENTROPY
                *
                loss_entropy

            )


        # ====================================================
        # BACKPROPAGATION
        # ====================================================

        if USE_AMP:

            scaler.scale(
                loss
            ).backward()


            scaler.unscale_(
                optimizer
            )


            torch.nn.utils.clip_grad_norm_(

                model.parameters(),

                1.0

            )


            scaler.step(
                optimizer
            )


            scaler.update()

        else:

            loss.backward()


            torch.nn.utils.clip_grad_norm_(

                model.parameters(),

                1.0

            )


            optimizer.step()


        # ====================================================
        # METRICS
        # ====================================================

        predictions = torch.argmax(

            logits,

            dim=1

        )


        correct = (

            predictions == labels

        ).sum().item()


        running_correct += correct

        running_total += labels.size(0)


        running_loss += (

            loss.item()

            *

            labels.size(0)

        )


        current_loss = (

            running_loss

            /

            running_total

        )


        current_accuracy = (

            running_correct

            /

            running_total

        )


        progress.set_postfix(

            loss=f"{current_loss:.4f}",

            accuracy=f"{current_accuracy:.4f}"

        )


    epoch_loss = (

        running_loss

        /

        running_total

    )


    epoch_accuracy = (

        running_correct

        /

        running_total

    )


    return (

        epoch_loss,

        epoch_accuracy

    )


# ============================================================
# 28. EVALUATION
# ============================================================

def evaluate(loader):

    """

    Validation/test mode.

    IMPORTANT:
    No task-sensitivity gradient is calculated here.

    """

    model.eval()


    running_loss = 0.0

    running_total = 0


    all_labels = []

    all_predictions = []


    with torch.no_grad():

        progress = tqdm(

            loader,

            desc="Evaluation",

            leave=False

        )


        for (

            rgb,

            depth,

            synthetic,

            labels

        ) in progress:


            rgb = rgb.to(

                DEVICE,

                non_blocking=True

            )


            depth = depth.to(

                DEVICE,

                non_blocking=True

            )


            synthetic = synthetic.to(

                DEVICE,

                non_blocking=True

            )


            labels = labels.to(

                DEVICE,

                non_blocking=True

            )


            with torch.autocast(

                device_type=DEVICE.type,

                dtype=torch.float16,

                enabled=USE_AMP

            ):

                output = model(

                    rgb,

                    depth,

                    synthetic,

                    sensitivity=None

                )


                logits = output[
                    "logits"
                ]


                loss = classification_loss(

                    logits,

                    labels

                )


            running_loss += (

                loss.item()

                *

                labels.size(0)

            )


            running_total += labels.size(0)


            predictions = torch.argmax(

                logits,

                dim=1

            )


            all_labels.extend(

                labels.cpu().numpy()

            )


            all_predictions.extend(

                predictions.cpu().numpy()

            )


    epoch_loss = (

        running_loss

        /

        running_total

    )


    accuracy = accuracy_score(

        all_labels,

        all_predictions

    )


    precision = precision_score(

        all_labels,

        all_predictions,

        average="macro",

        zero_division=0

    )


    recall = recall_score(

        all_labels,

        all_predictions,

        average="macro",

        zero_division=0

    )


    f1 = f1_score(

        all_labels,

        all_predictions,

        average="macro",

        zero_division=0

    )


    return (

        epoch_loss,

        accuracy,

        precision,

        recall,

        f1,

        all_labels,

        all_predictions

    )


# ============================================================
# 29. HISTORY
# ============================================================

history = {

    "train_loss": [],

    "train_accuracy": [],

    "val_loss": [],

    "val_accuracy": [],

    "val_precision": [],

    "val_recall": [],

    "val_f1": []

}


best_f1 = -1.0


# ============================================================
# 30. TRAIN
# ============================================================

print("\n")
print("=" * 70)
print("STARTING TRAINING")
print("=" * 70)


for epoch in range(

    NUM_EPOCHS

):

    print("\n")

    print(
        f"Epoch {epoch + 1}/{NUM_EPOCHS}"
    )


    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    train_loss, train_accuracy = (

        train_one_epoch()

    )


    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    (

        val_loss,

        val_accuracy,

        val_precision,

        val_recall,

        val_f1,

        _,

        _

    ) = evaluate(

        val_loader

    )


    scheduler.step()


    # --------------------------------------------------------
    # SAVE HISTORY
    # --------------------------------------------------------

    history[
        "train_loss"
    ].append(
        train_loss
    )


    history[
        "train_accuracy"
    ].append(
        train_accuracy
    )


    history[
        "val_loss"
    ].append(
        val_loss
    )


    history[
        "val_accuracy"
    ].append(
        val_accuracy
    )


    history[
        "val_precision"
    ].append(
        val_precision
    )


    history[
        "val_recall"
    ].append(
        val_recall
    )


    history[
        "val_f1"
    ].append(
        val_f1
    )


    # --------------------------------------------------------
    # RESULTS
    # --------------------------------------------------------

    print("\nResults:")

    print(
        f"Train Loss     : {train_loss:.4f}"
    )

    print(
        f"Train Accuracy : {train_accuracy:.4f}"
    )

    print(
        f"Val Loss       : {val_loss:.4f}"
    )

    print(
        f"Val Accuracy   : {val_accuracy:.4f}"
    )

    print(
        f"Val Precision  : {val_precision:.4f}"
    )

    print(
        f"Val Recall     : {val_recall:.4f}"
    )

    print(
        f"Val F1         : {val_f1:.4f}"
    )


    # --------------------------------------------------------
    # GPU MEMORY
    # --------------------------------------------------------

    if DEVICE.type == "cuda":

        allocated = (

            torch.cuda.memory_allocated(0)

            /

            (1024 ** 3)

        )


        reserved = (

            torch.cuda.memory_reserved(0)

            /

            (1024 ** 3)

        )


        print(

            f"GPU Memory     : "
            f"{allocated:.2f} GB allocated | "
            f"{reserved:.2f} GB reserved"

        )


    # --------------------------------------------------------
    # SAVE BEST MODEL
    # --------------------------------------------------------

    if val_f1 > best_f1:

        best_f1 = val_f1


        torch.save(

            {

                "model_state_dict":
                    model.state_dict(),

                "best_f1":
                    best_f1,

                "epoch":
                    epoch + 1,

                "class_names":
                    CLASS_NAMES,

                "alpha":
                    ALPHA

            },

            "best_FESR_ViT_Potato.pth"

        )


        print(
            ">>> BEST MODEL SAVED"
        )


# ============================================================
# 31. LOAD BEST MODEL
# ============================================================

print("\n")
print("=" * 70)
print("LOADING BEST MODEL")
print("=" * 70)


checkpoint = torch.load(

    "best_FESR_ViT_Potato.pth",

    map_location=DEVICE

)


model.load_state_dict(

    checkpoint[
        "model_state_dict"
    ]

)


print(

    "Best validation F1:",

    checkpoint[
        "best_f1"
    ]

)


# ============================================================
# 32. FINAL TEST
# ============================================================

(

    test_loss,

    test_accuracy,

    test_precision,

    test_recall,

    test_f1,

    y_true,

    y_pred

) = evaluate(

    test_loader

)


print("\n")
print("=" * 70)
print("FINAL TEST RESULTS")
print("=" * 70)


print(

    f"Test Loss     : "
    f"{test_loss:.4f}"

)


print(

    f"Accuracy      : "
    f"{test_accuracy:.4f}"

)


print(

    f"Precision     : "
    f"{test_precision:.4f}"

)


print(

    f"Recall        : "
    f"{test_recall:.4f}"

)


print(

    f"F1 Score      : "
    f"{test_f1:.4f}"

)


print("\n")


# ============================================================
# 33. CLASSIFICATION REPORT
# ============================================================

report = classification_report(

    y_true,

    y_pred,

    target_names=CLASS_NAMES,

    digits=4,

    zero_division=0

)


print("=" * 70)

print("CLASSIFICATION REPORT")

print("=" * 70)

print(report)


with open(

    "FESR_ViT_Classification_Report.txt",

    "w"

) as f:

    f.write(report)


# ============================================================
# 34. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_true,

    y_pred

)


print(
    "Confusion Matrix:"
)

print(cm)


plt.figure(

    figsize=(8, 6)

)


plt.imshow(

    cm,

    interpolation="nearest",

    cmap="Blues"

)


plt.title(
    "FESR-ViT Confusion Matrix"
)


plt.colorbar()


ticks = np.arange(
    NUM_CLASSES
)


plt.xticks(

    ticks,

    CLASS_NAMES,

    rotation=30

)


plt.yticks(

    ticks,

    CLASS_NAMES

)


threshold = (

    cm.max()

    /

    2

)


for i in range(

    NUM_CLASSES

):

    for j in range(

        NUM_CLASSES

    ):

        plt.text(

            j,

            i,

            str(cm[i, j]),

            ha="center",

            va="center",

            color=(

                "white"

                if cm[i, j] > threshold

                else "black"

            )

        )


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "Actual Class"
)


plt.tight_layout()


plt.savefig(

    "FESR_ViT_Confusion_Matrix.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 35. NORMALIZED CONFUSION MATRIX
# ============================================================

row_sums = cm.sum(

    axis=1,

    keepdims=True

)


row_sums[row_sums == 0] = 1


cm_norm = (

    cm.astype(np.float32)

    /

    row_sums

)


plt.figure(

    figsize=(8, 6)

)


plt.imshow(

    cm_norm,

    interpolation="nearest",

    cmap="Blues"

)


plt.title(

    "FESR-ViT Normalized Confusion Matrix"

)


plt.colorbar()


plt.xticks(

    ticks,

    CLASS_NAMES,

    rotation=30

)


plt.yticks(

    ticks,

    CLASS_NAMES

)


for i in range(

    NUM_CLASSES

):

    for j in range(

        NUM_CLASSES

    ):

        plt.text(

            j,

            i,

            f"{cm_norm[i,j]:.2f}",

            ha="center",

            va="center"

        )


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "Actual Class"
)


plt.tight_layout()


plt.savefig(

    "FESR_ViT_Normalized_Confusion_Matrix.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 36. ACCURACY CURVE
# ============================================================

plt.figure(

    figsize=(8, 6)

)


plt.plot(

    history["train_accuracy"],

    label="Training Accuracy",

    linewidth=2

)


plt.plot(

    history["val_accuracy"],

    label="Validation Accuracy",

    linewidth=2

)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.title(

    "FESR-ViT Accuracy"

)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    "FESR_ViT_Accuracy.png",

    dpi=300

)


plt.show()


# ============================================================
# 37. LOSS CURVE
# ============================================================

plt.figure(

    figsize=(8, 6)

)


plt.plot(

    history["train_loss"],

    label="Training Loss",

    linewidth=2

)


plt.plot(

    history["val_loss"],

    label="Validation Loss",

    linewidth=2

)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Loss"
)


plt.title(

    "FESR-ViT Loss"

)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    "FESR_ViT_Loss.png",

    dpi=300

)


plt.show()


# ============================================================
# 38. F1 CURVE
# ============================================================

plt.figure(

    figsize=(8, 6)

)


plt.plot(

    history["val_f1"],

    label="Validation F1",

    linewidth=2

)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "F1 Score"
)


plt.title(

    "FESR-ViT Validation F1"

)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    "FESR_ViT_F1.png",

    dpi=300

)


plt.show()


# ============================================================
# 39. PRECISION / RECALL
# ============================================================

plt.figure(

    figsize=(8, 6)

)


plt.plot(

    history["val_precision"],

    label="Validation Precision",

    linewidth=2

)


plt.plot(

    history["val_recall"],

    label="Validation Recall",

    linewidth=2

)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Score"
)


plt.title(

    "FESR-ViT Precision and Recall"

)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    "FESR_ViT_Precision_Recall.png",

    dpi=300

)


plt.show()


# ============================================================
# 40. SAVE NUMERICAL RESULTS
# ============================================================

with open(

    "FESR_ViT_Final_Results.txt",

    "w"

) as f:

    f.write(
        "FESR-ViT Potato Classification\n"
    )

    f.write(
        "==============================\n\n"
    )

    f.write(

        f"Accuracy  : "
        f"{test_accuracy:.6f}\n"

    )

    f.write(

        f"Precision : "
        f"{test_precision:.6f}\n"

    )

    f.write(

        f"Recall    : "
        f"{test_recall:.6f}\n"

    )

    f.write(

        f"F1 Score  : "
        f"{test_f1:.6f}\n\n"

    )

    f.write(
        "Confusion Matrix:\n"
    )

    f.write(
        str(cm)
    )


# ============================================================
# 41. FINAL OUTPUT
# ============================================================

print("\n")
print("=" * 70)
print("TRAINING COMPLETED SUCCESSFULLY")
print("=" * 70)


print(
    "\nFinal Test Accuracy:",
    f"{test_accuracy * 100:.2f}%"
)


print(
    "Final Test Precision:",
    f"{test_precision * 100:.2f}%"
)


print(
    "Final Test Recall:",
    f"{test_recall * 100:.2f}%"
)


print(
    "Final Test F1:",
    f"{test_f1 * 100:.2f}%"
)


if DEVICE.type == "cuda":

    print(

        "\nGPU used:",

        torch.cuda.get_device_name(0)

    )


print("\nGenerated files:")

print(
    "best_FESR_ViT_Potato.pth"
)

print(
    "FESR_ViT_Classification_Report.txt"
)

print(
    "FESR_ViT_Final_Results.txt"
)

print(
    "FESR_ViT_Confusion_Matrix.png"
)

print(
    "FESR_ViT_Normalized_Confusion_Matrix.png"
)

print(
    "FESR_ViT_Accuracy.png"
)

print(
    "FESR_ViT_Loss.png"
)

print(
    "FESR_ViT_F1.png"
)

print(
    "FESR_ViT_Precision_Recall.png"
)

print("\nDone.")
```

```python
# ============================================================
# FERL-ViT: FEATURE ENERGY-REDISTRIBUTION LEARNING
# MAIZE CLASSIFICATION
#
# Dataset:
#   Maize ZIP
#
# Split:
#   70% Train
#   15% Validation
#   15% Test
#
# Epochs:
#   20
#
# GPU:
#   Automatically uses CUDA / Tesla T4 if available
#
# Architecture:
#
# RGB ───────────────┐
#                    │
# Pseudo-Depth ──────┼──> Feature Extraction
#                    │
# Synthetic ────────┘
#                         ↓
#                    Feature Fusion
#                         ↓
#                  Energy Estimation
#                         +
#                  Task Sensitivity
#                         ↓
#              Energy-Sensitivity Integration
#                         ↓
#               Adaptive Redistribution
#                         ↓
#                Balanced Representation
#                         ↓
#                        ViT
#                         ↓
#                     Classifier
#                         ↓
#                    Prediction
#
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import os
import glob
import zipfile
import random
import warnings

import numpy as np

import matplotlib.pyplot as plt

from PIL import Image
from PIL import ImageOps
from PIL import ImageEnhance

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset
from torch.utils.data import DataLoader
from torch.utils.data import WeightedRandomSampler

from torchvision import transforms

from sklearn.model_selection import train_test_split

from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

from tqdm.auto import tqdm


warnings.filterwarnings("ignore")


# ============================================================
# 2. CONFIGURATION
# ============================================================

IMAGE_SIZE = 224

BATCH_SIZE = 16

NUM_EPOCHS = 20

LEARNING_RATE = 1e-4

WEIGHT_DECAY = 1e-4

NUM_CLASSES = 4

SEED = 42

NUM_WORKERS = 2


# ============================================================
# 3. FERL PARAMETERS
# ============================================================

# Alpha controls the contribution of task sensitivity
# relative to raw feature energy.

ALPHA = 0.30

EPSILON = 1e-8

LAMBDA_REDISTRIBUTION = 0.10

LAMBDA_CONSISTENCY = 0.05

LAMBDA_ENTROPY = 0.01


# ============================================================
# 4. CLASS NAMES
# ============================================================

CLASS_NAMES = [

    "Cercospora Leaf Spot",

    "Common Rust",

    "Healthy",

    "Northern Leaf Blight"

]


# ============================================================
# 5. REPRODUCIBILITY
# ============================================================

def set_seed(seed):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():

        torch.cuda.manual_seed(seed)

        torch.cuda.manual_seed_all(seed)


set_seed(SEED)


# ============================================================
# 6. GPU CONFIGURATION
# ============================================================

print("=" * 70)
print("GPU CONFIGURATION")
print("=" * 70)


if torch.cuda.is_available():

    DEVICE = torch.device("cuda")

    print("GPU DETECTED")

    print(
        "GPU:",
        torch.cuda.get_device_name(0)
    )

    print(
        "CUDA:",
        torch.version.cuda
    )

    print(
        "Device:",
        DEVICE
    )

    torch.backends.cudnn.benchmark = True

    torch.backends.cuda.matmul.allow_tf32 = True

    torch.backends.cudnn.allow_tf32 = True

    USE_AMP = True

else:

    DEVICE = torch.device("cpu")

    USE_AMP = False

    print("GPU NOT DETECTED")

    print(
        "Device:",
        DEVICE
    )


# ============================================================
# 7. AUTOMATICALLY FIND DATASET ZIP
# ============================================================

print("\n")
print("=" * 70)
print("SEARCHING FOR MAIZE DATASET")
print("=" * 70)


search_locations = [

    "/mnt/data",

    "/content",

    "/kaggle/input",

    "/workspace"

]


possible_files = []


for location in search_locations:

    if not os.path.exists(location):

        continue


    # ZIP files

    possible_files.extend(

        glob.glob(

            os.path.join(

                location,

                "**",

                "*.zip"

            ),

            recursive=True

        )

    )


# Remove duplicates

possible_files = list(

    dict.fromkeys(

        possible_files

    )

)


print("\nZIP files found:")


for file in possible_files:

    print(
        "  ",
        file
    )


# ============================================================
# 8. SELECT MAIZE ZIP
# ============================================================

ZIP_PATH = None


# Prefer filename containing maize

for file in possible_files:

    filename = os.path.basename(
        file
    ).lower()


    if "maize" in filename:

        ZIP_PATH = file

        break


# If only one ZIP exists

if ZIP_PATH is None:

    if len(possible_files) == 1:

        ZIP_PATH = possible_files[0]


# ============================================================
# 9. CHECK DATASET
# ============================================================

if ZIP_PATH is None:

    print("\n")
    print("=" * 70)
    print("ERROR")
    print("=" * 70)

    print(
        "No ZIP file containing the Maize dataset was found."
    )

    print("\nFiles visible in /mnt/data:")

    for root, dirs, files in os.walk(
        "/mnt/data"
    ):

        for file in files:

            print(

                os.path.join(

                    root,

                    file

                )

            )


    raise FileNotFoundError(

        "Maize dataset ZIP could not be found."

    )


print("\n")
print("=" * 70)
print("DATASET FOUND")
print("=" * 70)

print(
    "ZIP PATH:"
)

print(
    ZIP_PATH
)


# ============================================================
# 10. EXTRACT ZIP
# ============================================================

EXTRACT_PATH = "/mnt/data/Maize_extracted"


if not os.path.exists(
    EXTRACT_PATH
):

    print("\nExtracting dataset...")

    with zipfile.ZipFile(

        ZIP_PATH,

        "r"

    ) as zip_ref:

        zip_ref.extractall(

            EXTRACT_PATH

        )

    print(
        "Extraction completed."
    )

else:

    print(
        "\nDataset already extracted."
    )


# ============================================================
# 11. FIND DATASET ROOT
# ============================================================

def find_dataset_root(
    base_path
):

    candidates = []


    for root, dirs, files in os.walk(
        base_path
    ):

        names = [

            d.lower()

            for d in dirs

        ]


        has_healthy = any(

            "healthy" in d

            for d in names

        )


        has_rust = any(

            "rust" in d

            for d in names

        )


        has_blight = any(

            "blight" in d

            for d in names

        )


        has_spot = any(

            (

                "spot" in d

                or

                "cercospora" in d

                or

                "gray" in d

            )

            for d in names

        )


        if (

            has_healthy

            and

            has_rust

            and

            has_blight

            and

            has_spot

        ):

            candidates.append(

                root

            )


    return candidates


roots = find_dataset_root(

    EXTRACT_PATH

)


if len(roots) == 0:

    print(
        "\nDataset root could not be automatically detected."
    )

    print(
        "\nExtracted directory structure:"
    )


    for root, dirs, files in os.walk(

        EXTRACT_PATH

    ):

        level = root.replace(

            EXTRACT_PATH,

            ""

        ).count(

            os.sep

        )


        if level <= 3:

            print(

                "  " * level

                +

                os.path.basename(root)

                +

                "/"

            )


            for d in dirs:

                print(

                    "  " * (level + 1)

                    +

                    d

                    +

                    "/"

                )


    raise RuntimeError(

        "Could not locate the four Maize classes."

    )


DATA_ROOT = roots[0]


print("\n")
print("=" * 70)
print("DATASET ROOT")
print("=" * 70)

print(
    DATA_ROOT
)


print("\nClass folders:")


for folder in os.listdir(
    DATA_ROOT
):

    folder_path = os.path.join(

        DATA_ROOT,

        folder

    )


    if os.path.isdir(
        folder_path
    ):

        print(
            "  ",
            folder
        )


# ============================================================
# 12. CLASS LABEL MAPPING
# ============================================================

def get_class_label(
    folder_name
):

    name = folder_name.lower()


    # Class 0
    if (

        "cercospora" in name

        or

        "gray_leaf_spot" in name

        or

        "gray leaf spot" in name

        or

        (

            "gray" in name

            and

            "spot" in name

        )

    ):

        return 0


    # Class 1
    if (

        "common_rust" in name

        or

        "common rust" in name

    ):

        return 1


    # Class 2
    if "healthy" in name:

        return 2


    # Class 3
    if (

        "northern_leaf_blight" in name

        or

        "northern leaf blight" in name

        or

        (

            "northern" in name

            and

            "blight" in name

        )

    ):

        return 3


    return None


# ============================================================
# 13. COLLECT IMAGES
# ============================================================

IMAGE_EXTENSIONS = (

    ".jpg",

    ".jpeg",

    ".png",

    ".bmp",

    ".webp",

    ".JPG",

    ".JPEG",

    ".PNG",

    ".BMP",

    ".WEBP"

)


image_paths = []

labels = []


for class_folder in os.listdir(
    DATA_ROOT
):

    folder_path = os.path.join(

        DATA_ROOT,

        class_folder

    )


    if not os.path.isdir(
        folder_path
    ):

        continue


    label = get_class_label(

        class_folder

    )


    if label is None:

        continue


    for root, dirs, files in os.walk(

        folder_path

    ):

        for file in files:

            if file.endswith(

                IMAGE_EXTENSIONS

            ):

                image_paths.append(

                    os.path.join(

                        root,

                        file

                    )

                )

                labels.append(

                    label

                )


# ============================================================
# 14. DATASET VERIFICATION
# ============================================================

print("\n")
print("=" * 70)
print("DATASET VERIFICATION")
print("=" * 70)


print(

    "Total images:",

    len(image_paths)

)


if len(image_paths) == 0:

    raise RuntimeError(

        "No images found."

    )


for i, class_name in enumerate(

    CLASS_NAMES

):

    count = labels.count(i)


    print(

        f"{class_name}: {count}"

    )


# ============================================================
# 15. 70/15/15 STRATIFIED SPLIT
# ============================================================

train_paths, temp_paths, train_labels, temp_labels = (

    train_test_split(

        image_paths,

        labels,

        test_size=0.30,

        stratify=labels,

        random_state=SEED

    )

)


val_paths, test_paths, val_labels, test_labels = (

    train_test_split(

        temp_paths,

        temp_labels,

        test_size=0.50,

        stratify=temp_labels,

        random_state=SEED

    )

)


print("\n")
print("=" * 70)
print("DATA SPLIT")
print("=" * 70)


print(

    f"Training   : {len(train_paths)} "

    f"({len(train_paths) / len(image_paths) * 100:.2f}%)"

)


print(

    f"Validation : {len(val_paths)} "

    f"({len(val_paths) / len(image_paths) * 100:.2f}%)"

)


print(

    f"Testing    : {len(test_paths)} "

    f"({len(test_paths) / len(image_paths) * 100:.2f}%)"

)


# ============================================================
# 16. TRANSFORMS
# ============================================================

train_transform = transforms.Compose([

    transforms.Resize(

        (IMAGE_SIZE, IMAGE_SIZE)

    ),

    transforms.RandomHorizontalFlip(

        p=0.5

    ),

    transforms.RandomVerticalFlip(

        p=0.2

    ),

    transforms.RandomRotation(

        15

    ),

    transforms.ColorJitter(

        brightness=0.20,

        contrast=0.20,

        saturation=0.20,

        hue=0.05

    ),

    transforms.ToTensor(),

    transforms.Normalize(

        [0.485, 0.456, 0.406],

        [0.229, 0.224, 0.225]

    )

])


test_transform = transforms.Compose([

    transforms.Resize(

        (IMAGE_SIZE, IMAGE_SIZE)

    ),

    transforms.ToTensor(),

    transforms.Normalize(

        [0.485, 0.456, 0.406],

        [0.229, 0.224, 0.225]

    )

])


# ============================================================
# 17. PSEUDO-DEPTH GENERATION
# ============================================================

def create_pseudo_depth(
    image
):

    gray = ImageOps.grayscale(

        image

    )


    gray = gray.resize(

        (

            IMAGE_SIZE,

            IMAGE_SIZE

        )

    )


    gray = ImageEnhance.Contrast(

        gray

    ).enhance(

        1.5

    )


    depth = np.asarray(

        gray,

        dtype=np.float32

    ) / 255.0


    depth = torch.from_numpy(

        depth

    )


    depth = depth.unsqueeze(

        0

    )


    return depth


# ============================================================
# 18. SYNTHETIC IMAGE GENERATION
# ============================================================

def create_synthetic_image(
    image
):

    img = image.copy()


    choice = random.randint(

        0,

        4

    )


    if choice == 0:

        img = ImageEnhance.Brightness(

            img

        ).enhance(

            random.uniform(

                0.75,

                1.25

            )

        )


    elif choice == 1:

        img = ImageEnhance.Contrast(

            img

        ).enhance(

            random.uniform(

                0.75,

                1.25

            )

        )


    elif choice == 2:

        img = ImageEnhance.Color(

            img

        ).enhance(

            random.uniform(

                0.70,

                1.30

            )

        )


    elif choice == 3:

        img = ImageOps.mirror(

            img

        )


    else:

        img = img.rotate(

            random.uniform(

                -15,

                15

            )

        )


    return img


# ============================================================
# 19. DATASET CLASS
# ============================================================

class MaizeDataset(
    Dataset
):

    def __init__(

        self,

        paths,

        labels,

        transform

    ):

        self.paths = paths

        self.labels = labels

        self.transform = transform


    def __len__(
        self
    ):

        return len(

            self.paths

        )


    def __getitem__(
        self,

        index

    ):

        path = self.paths[index]

        label = self.labels[index]


        image = Image.open(

            path

        ).convert(

            "RGB"

        )


        original = image.copy()


        # ----------------------------------------------------
        # RGB
        # ----------------------------------------------------

        rgb = self.transform(

            image

        )


        # ----------------------------------------------------
        # PSEUDO DEPTH
        # ----------------------------------------------------

        depth = create_pseudo_depth(

            original

        )


        # ----------------------------------------------------
        # SYNTHETIC
        # ----------------------------------------------------

        synthetic_image = (

            create_synthetic_image(

                original

            )

        )


        synthetic = self.transform(

            synthetic_image

        )


        return (

            rgb,

            depth,

            synthetic,

            torch.tensor(

                label,

                dtype=torch.long

            )

        )


# ============================================================
# 20. CREATE DATASETS
# ============================================================

train_dataset = MaizeDataset(

    train_paths,

    train_labels,

    train_transform

)


val_dataset = MaizeDataset(

    val_paths,

    val_labels,

    test_transform

)


test_dataset = MaizeDataset(

    test_paths,

    test_labels,

    test_transform

)


# ============================================================
# 21. CLASS BALANCING
# ============================================================

class_counts = np.bincount(

    train_labels,

    minlength=NUM_CLASSES

)


class_weights = (

    len(train_labels)

    /

    (

        NUM_CLASSES

        *

        class_counts

    )

)


sample_weights = [

    class_weights[label]

    for label in train_labels

]


sampler = WeightedRandomSampler(

    torch.tensor(

        sample_weights,

        dtype=torch.double

    ),

    num_samples=len(

        sample_weights

    ),

    replacement=True

)


# ============================================================
# 22. DATALOADERS
# ============================================================

train_loader = DataLoader(

    train_dataset,

    batch_size=BATCH_SIZE,

    sampler=sampler,

    num_workers=NUM_WORKERS,

    pin_memory=(

        DEVICE.type == "cuda"

    ),

    persistent_workers=(

        NUM_WORKERS > 0

    )

)


val_loader = DataLoader(

    val_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=(

        DEVICE.type == "cuda"

    ),

    persistent_workers=(

        NUM_WORKERS > 0

    )

)


test_loader = DataLoader(

    test_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=(

        DEVICE.type == "cuda"

    ),

    persistent_workers=(

        NUM_WORKERS > 0

    )

)


# ============================================================
# 23. FEATURE ENCODER
# ============================================================

class FeatureEncoder(
    nn.Module
):

    def __init__(

        self,

        input_channels,

        embed_dim=192

    ):

        super().__init__()


        self.encoder = nn.Sequential(

            nn.Conv2d(

                input_channels,

                64,

                kernel_size=3,

                stride=2,

                padding=1

            ),

            nn.BatchNorm2d(

                64

            ),

            nn.GELU(),


            nn.Conv2d(

                64,

                96,

                kernel_size=3,

                stride=2,

                padding=1

            ),

            nn.BatchNorm2d(

                96

            ),

            nn.GELU(),


            nn.Conv2d(

                96,

                128,

                kernel_size=3,

                stride=2,

                padding=1

            ),

            nn.BatchNorm2d(

                128

            ),

            nn.GELU(),


            nn.Conv2d(

                128,

                embed_dim,

                kernel_size=3,

                stride=2,

                padding=1

            ),

            nn.BatchNorm2d(

                embed_dim

            ),

            nn.GELU()

        )


    def forward(
        self,
        x
    ):

        x = self.encoder(

            x

        )


        x = x.flatten(

            2

        )


        x = x.transpose(

            1,

            2

        )


        return x


# ============================================================
# 24. FERL-VIT MODEL
# ============================================================

class FERLViT(
    nn.Module
):

    def __init__(

        self,

        num_classes=4,

        embed_dim=192,

        depth=6,

        heads=6

    ):

        super().__init__()


        # ----------------------------------------------------
        # THREE MODALITY ENCODERS
        # ----------------------------------------------------

        self.rgb_encoder = FeatureEncoder(

            3,

            embed_dim

        )


        self.depth_encoder = FeatureEncoder(

            1,

            embed_dim

        )


        self.synthetic_encoder = FeatureEncoder(

            3,

            embed_dim

        )


        # ----------------------------------------------------
        # FEATURE FUSION
        # ----------------------------------------------------

        self.fusion = nn.Linear(

            embed_dim * 3,

            embed_dim

        )


        # ----------------------------------------------------
        # CLASS TOKEN
        # ----------------------------------------------------

        self.cls_token = nn.Parameter(

            torch.zeros(

                1,

                1,

                embed_dim

            )

        )


        # 224 / 16 = 14
        # 14 x 14 = 196 patches
        # + 1 CLS = 197

        self.position_embedding = nn.Parameter(

            torch.zeros(

                1,

                197,

                embed_dim

            )

        )


        # ----------------------------------------------------
        # TRANSFORMER
        # ----------------------------------------------------

        encoder_layer = (

            nn.TransformerEncoderLayer(

                d_model=embed_dim,

                nhead=heads,

                dim_feedforward=embed_dim * 4,

                dropout=0.1,

                activation="gelu",

                batch_first=True,

                norm_first=True

            )

        )


        self.transformer = (

            nn.TransformerEncoder(

                encoder_layer,

                num_layers=depth

            )

        )


        self.norm = nn.LayerNorm(

            embed_dim

        )


        # ----------------------------------------------------
        # CLASSIFIER
        # ----------------------------------------------------

        self.classifier = nn.Linear(

            embed_dim,

            num_classes

        )


        nn.init.trunc_normal_(

            self.cls_token,

            std=0.02

        )


        nn.init.trunc_normal_(

            self.position_embedding,

            std=0.02

        )


    # ========================================================
    # ENERGY
    # ========================================================

    def calculate_energy(

        self,

        features

    ):

        return torch.mean(

            features ** 2,

            dim=2

        )


    # ========================================================
    # ENERGY DISTRIBUTION
    # ========================================================

    def calculate_energy_distribution(

        self,

        features

    ):

        energy = (

            self.calculate_energy(

                features

            )

        )


        return (

            energy

            /

            (

                energy.sum(

                    dim=1,

                    keepdim=True

                )

                +

                EPSILON

            )

        )


    # ========================================================
    # ENERGY-SENSITIVITY INTEGRATION
    # ========================================================

    def energy_sensitivity_integration(

        self,

        features,

        sensitivity

    ):

        energy = (

            self.calculate_energy_distribution(

                features

            )

        )


        sensitivity = (

            sensitivity

            /

            (

                sensitivity.sum(

                    dim=1,

                    keepdim=True

                )

                +

                EPSILON

            )

        )


        target = (

            (1 - ALPHA) * energy

            +

            ALPHA * sensitivity

        )


        target = (

            target

            /

            (

                target.sum(

                    dim=1,

                    keepdim=True

                )

                +

                EPSILON

            )

        )


        return (

            energy,

            sensitivity,

            target

        )


    # ========================================================
    # ADAPTIVE REDISTRIBUTION
    # ========================================================

    def adaptive_redistribution(

        self,

        features,

        sensitivity

    ):

        (

            energy,

            sensitivity_distribution,

            target_distribution

        ) = (

            self.energy_sensitivity_integration(

                features,

                sensitivity

            )

        )


        scale = torch.sqrt(

            (

                target_distribution

                +

                EPSILON

            )

            /

            (

                energy

                +

                EPSILON

            )

        )


        scale = torch.clamp(

            scale,

            0.25,

            4.0

        )


        balanced_features = (

            features

            *

            scale.unsqueeze(

                2

            )

        )


        return (

            balanced_features,

            energy,

            sensitivity_distribution,

            target_distribution

        )


    # ========================================================
    # FORWARD
    # ========================================================

    def forward(

        self,

        rgb,

        depth,

        synthetic,

        sensitivity=None

    ):

        # ----------------------------------------------------
        # FEATURE EXTRACTION
        # ----------------------------------------------------

        F_R = self.rgb_encoder(

            rgb

        )


        F_D = self.depth_encoder(

            depth

        )


        F_S = self.synthetic_encoder(

            synthetic

        )


        # ----------------------------------------------------
        # MATCH TOKEN COUNT
        # ----------------------------------------------------

        n = min(

            F_R.shape[1],

            F_D.shape[1],

            F_S.shape[1]

        )


        F_R = F_R[:, :n, :]

        F_D = F_D[:, :n, :]

        F_S = F_S[:, :n, :]


        # ----------------------------------------------------
        # FUSION
        # ----------------------------------------------------

        fused_features = torch.cat(

            [

                F_R,

                F_D,

                F_S

            ],

            dim=-1

        )


        fused_features = self.fusion(

            fused_features

        )


        # ----------------------------------------------------
        # IMPORTANT FIX
        #
        # retain_grad() is ONLY allowed when:
        #
        # 1. Training
        # 2. Gradients enabled
        # 3. Tensor requires gradient
        #
        # This prevents:
        #
        # RuntimeError:
        # can't retain_grad on Tensor that
        # has requires_grad=False
        #
        # ----------------------------------------------------

        if (

            self.training

            and

            torch.is_grad_enabled()

            and

            fused_features.requires_grad

        ):

            fused_features.retain_grad()


        # ----------------------------------------------------
        # FERL REDISTRIBUTION
        # ----------------------------------------------------

        if sensitivity is not None:

            (

                balanced_features,

                energy_distribution,

                sensitivity_distribution,

                target_distribution

            ) = (

                self.adaptive_redistribution(

                    fused_features,

                    sensitivity

                )

            )

        else:

            energy_distribution = (

                self.calculate_energy_distribution(

                    fused_features

                )

            )


            balanced_features = (

                fused_features

            )


            sensitivity_distribution = (

                energy_distribution

            )


            target_distribution = (

                energy_distribution

            )


        # ----------------------------------------------------
        # CLS TOKEN
        # ----------------------------------------------------

        B = balanced_features.shape[0]


        cls = self.cls_token.expand(

            B,

            -1,

            -1

        )


        tokens = torch.cat(

            [

                cls,

                balanced_features

            ],

            dim=1

        )


        # ----------------------------------------------------
        # POSITION EMBEDDING
        # ----------------------------------------------------

        tokens = (

            tokens

            +

            self.position_embedding[

                :,

                :tokens.shape[1],

                :

            ]

        )


        # ----------------------------------------------------
        # VIT
        # ----------------------------------------------------

        encoded = self.transformer(

            tokens

        )


        encoded = self.norm(

            encoded

        )


        representation = encoded[

            :,

            0

        ]


        # ----------------------------------------------------
        # CLASSIFIER
        # ----------------------------------------------------

        logits = self.classifier(

            representation

        )


        return {

            "logits":
                logits,

            "features":
                fused_features,

            "balanced_features":
                balanced_features,

            "energy_distribution":
                energy_distribution,

            "sensitivity_distribution":
                sensitivity_distribution,

            "target_distribution":
                target_distribution,

            "synthetic_features":
                F_S

        }


# ============================================================
# 25. CREATE MODEL
# ============================================================

model = FERLViT(

    num_classes=NUM_CLASSES,

    embed_dim=192,

    depth=6,

    heads=6

).to(

    DEVICE

)


total_params = sum(

    p.numel()

    for p in model.parameters()

)


trainable_params = sum(

    p.numel()

    for p in model.parameters()

    if p.requires_grad

)


print("\n")
print("=" * 70)
print("MODEL")
print("=" * 70)


print(

    f"Total parameters: "
    f"{total_params:,}"

)


print(

    f"Trainable parameters: "
    f"{trainable_params:,}"

)


# ============================================================
# 26. CLASSIFICATION LOSS
# ============================================================

classification_loss = (

    nn.CrossEntropyLoss(

        label_smoothing=0.1

    )

)


# ============================================================
# 27. KL DIVERGENCE
# ============================================================

def kl_divergence(

    p,

    q

):

    p = torch.clamp(

        p,

        min=EPSILON

    )


    q = torch.clamp(

        q,

        min=EPSILON

    )


    return torch.mean(

        torch.sum(

            p *

            torch.log(

                p / q

            ),

            dim=1

        )

    )


# ============================================================
# 28. JS DIVERGENCE
# ============================================================

def js_divergence(

    p,

    q

):

    m = 0.5 * (

        p + q

    )


    return (

        0.5 *

        (

            kl_divergence(

                p,

                m

            )

            +

            kl_divergence(

                q,

                m

            )

        )

    )


# ============================================================
# 29. ENTROPY REGULARIZATION
# ============================================================

def entropy_regularization(

    p

):

    p = torch.clamp(

        p,

        min=EPSILON

    )


    entropy = -torch.sum(

        p *

        torch.log(p),

        dim=1

    )


    return -torch.mean(

        entropy

    )


# ============================================================
# 30. OPTIMIZER
# ============================================================

optimizer = torch.optim.AdamW(

    model.parameters(),

    lr=LEARNING_RATE,

    weight_decay=WEIGHT_DECAY

)


scheduler = (

    torch.optim.lr_scheduler.CosineAnnealingLR(

        optimizer,

        T_max=NUM_EPOCHS

    )

)


# ============================================================
# 31. AMP SCALER
# ============================================================

if USE_AMP:

    scaler = torch.amp.GradScaler(

        "cuda"

    )

else:

    scaler = None


# ============================================================
# 32. TRAIN ONE EPOCH
# ============================================================

def train_one_epoch():

    model.train()


    total_loss = 0.0

    total_correct = 0

    total_samples = 0


    progress = tqdm(

        train_loader,

        desc="Training",

        leave=True

    )


    for (

        rgb,

        depth,

        synthetic,

        labels

    ) in progress:


        rgb = rgb.to(

            DEVICE,

            non_blocking=True

        )


        depth = depth.to(

            DEVICE,

            non_blocking=True

        )


        synthetic = synthetic.to(

            DEVICE,

            non_blocking=True

        )


        labels = labels.to(

            DEVICE,

            non_blocking=True

        )


        optimizer.zero_grad(

            set_to_none=True

        )


        # ====================================================
        # FIRST FORWARD PASS
        # ====================================================

        with torch.autocast(

            device_type=DEVICE.type,

            dtype=torch.float16,

            enabled=USE_AMP

        ):

            first_output = model(

                rgb,

                depth,

                synthetic,

                sensitivity=None

            )


            first_logits = (

                first_output["logits"]

            )


            first_loss = (

                classification_loss(

                    first_logits,

                    labels

                )

            )


        # ====================================================
        # TASK SENSITIVITY
        #
        # S_i =
        # sqrt(mean((dL/dF_i)^2))
        # ====================================================

        features = first_output[

            "features"

        ]


        if USE_AMP:

            scaled_loss = (

                scaler.scale(

                    first_loss

                )

            )


            gradients = torch.autograd.grad(

                scaled_loss,

                features,

                retain_graph=True,

                create_graph=False,

                allow_unused=True

            )[0]


            if gradients is not None:

                gradients = (

                    gradients

                    /

                    scaler.get_scale()

                )

        else:

            gradients = torch.autograd.grad(

                first_loss,

                features,

                retain_graph=True,

                create_graph=False,

                allow_unused=True

            )[0]


        if gradients is None:

            sensitivity = torch.ones(

                features.shape[0],

                features.shape[1],

                device=DEVICE

            )

        else:

            sensitivity = torch.sqrt(

                torch.mean(

                    gradients.detach() ** 2,

                    dim=2

                )

                +

                EPSILON

            )


        # ====================================================
        # SECOND FORWARD PASS
        # ====================================================

        with torch.autocast(

            device_type=DEVICE.type,

            dtype=torch.float16,

            enabled=USE_AMP

        ):

            output = model(

                rgb,

                depth,

                synthetic,

                sensitivity=sensitivity

            )


            logits = output[

                "logits"

            ]


            # Classification loss

            loss_cls = (

                classification_loss(

                    logits,

                    labels

                )

            )


            # Energy

            energy = output[

                "energy_distribution"

            ]


            # Target

            target = output[

                "target_distribution"

            ]


            # Redistribution loss

            loss_red = kl_divergence(

                target,

                energy

            )


            # Synthetic consistency

            synthetic_features = output[

                "synthetic_features"

            ]


            synthetic_energy = (

                model.calculate_energy_distribution(

                    synthetic_features

                )

            )


            loss_consistency = js_divergence(

                energy,

                synthetic_energy

            )


            # Entropy

            loss_entropy = entropy_regularization(

                target

            )


            # TOTAL LOSS

            loss = (

                loss_cls

                +

                LAMBDA_REDISTRIBUTION

                * loss_red

                +

                LAMBDA_CONSISTENCY

                * loss_consistency

                +

                LAMBDA_ENTROPY

                * loss_entropy

            )


        # ====================================================
        # BACKPROPAGATION
        # ====================================================

        if USE_AMP:

            scaler.scale(

                loss

            ).backward()


            scaler.unscale_(

                optimizer

            )


            torch.nn.utils.clip_grad_norm_(

                model.parameters(),

                1.0

            )


            scaler.step(

                optimizer

            )


            scaler.update()

        else:

            loss.backward()


            torch.nn.utils.clip_grad_norm_(

                model.parameters(),

                1.0

            )


            optimizer.step()


        # ====================================================
        # METRICS
        # ====================================================

        predictions = torch.argmax(

            logits,

            dim=1

        )


        correct = (

            predictions == labels

        ).sum().item()


        total_correct += correct

        total_samples += labels.size(0)


        total_loss += (

            loss.item()

            *

            labels.size(0)

        )


        progress.set_postfix(

            loss=(

                total_loss

                /

                total_samples

            ),

            accuracy=(

                total_correct

                /

                total_samples

            )

        )


    return (

        total_loss / total_samples,

        total_correct / total_samples

    )


# ============================================================
# 33. EVALUATION
# ============================================================

def evaluate(

    loader

):

    model.eval()


    total_loss = 0.0

    total_samples = 0


    all_labels = []

    all_predictions = []


    # --------------------------------------------------------
    # IMPORTANT:
    #
    # Evaluation uses no gradients.
    # Therefore the model NEVER calls retain_grad()
    # because the forward function checks:
    #
    # self.training
    # torch.is_grad_enabled()
    # fused_features.requires_grad
    #
    # --------------------------------------------------------

    with torch.no_grad():

        for (

            rgb,

            depth,

            synthetic,

            labels

        ) in tqdm(

            loader,

            desc="Evaluation",

            leave=False

        ):


            rgb = rgb.to(

                DEVICE,

                non_blocking=True

            )


            depth = depth.to(

                DEVICE,

                non_blocking=True

            )


            synthetic = synthetic.to(

                DEVICE,

                non_blocking=True

            )


            labels = labels.to(

                DEVICE,

                non_blocking=True

            )


            with torch.autocast(

                device_type=DEVICE.type,

                dtype=torch.float16,

                enabled=USE_AMP

            ):

                output = model(

                    rgb,

                    depth,

                    synthetic,

                    sensitivity=None

                )


                logits = output[

                    "logits"

                ]


                loss = classification_loss(

                    logits,

                    labels

                )


            total_loss += (

                loss.item()

                *

                labels.size(0)

            )


            total_samples += labels.size(0)


            predictions = torch.argmax(

                logits,

                dim=1

            )


            all_labels.extend(

                labels.cpu().numpy()

            )


            all_predictions.extend(

                predictions.cpu().numpy()

            )


    avg_loss = (

        total_loss

        /

        total_samples

    )


    accuracy = accuracy_score(

        all_labels,

        all_predictions

    )


    precision = precision_score(

        all_labels,

        all_predictions,

        average="macro",

        zero_division=0

    )


    recall = recall_score(

        all_labels,

        all_predictions,

        average="macro",

        zero_division=0

    )


    f1 = f1_score(

        all_labels,

        all_predictions,

        average="macro",

        zero_division=0

    )


    return (

        avg_loss,

        accuracy,

        precision,

        recall,

        f1,

        all_labels,

        all_predictions

    )


# ============================================================
# 34. HISTORY
# ============================================================

history = {

    "train_loss": [],

    "train_accuracy": [],

    "val_loss": [],

    "val_accuracy": [],

    "val_precision": [],

    "val_recall": [],

    "val_f1": []

}


best_f1 = -1.0


# ============================================================
# 35. TRAINING
# ============================================================

print("\n")
print("=" * 70)
print("STARTING 20-EPOCH TRAINING")
print("=" * 70)


for epoch in range(

    NUM_EPOCHS

):

    print("\n")

    print(

        "=" * 70

    )

    print(

        f"EPOCH {epoch + 1}/{NUM_EPOCHS}"

    )

    print(

        "=" * 70

    )


    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    train_loss, train_accuracy = (

        train_one_epoch()

    )


    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    (

        val_loss,

        val_accuracy,

        val_precision,

        val_recall,

        val_f1,

        _,

        _

    ) = evaluate(

        val_loader

    )


    # --------------------------------------------------------
    # LR SCHEDULER
    # --------------------------------------------------------

    scheduler.step()


    # --------------------------------------------------------
    # STORE
    # --------------------------------------------------------

    history[

        "train_loss"

    ].append(

        train_loss

    )


    history[

        "train_accuracy"

    ].append(

        train_accuracy

    )


    history[

        "val_loss"

    ].append(

        val_loss

    )


    history[

        "val_accuracy"

    ].append(

        val_accuracy

    )


    history[

        "val_precision"

    ].append(

        val_precision

    )


    history[

        "val_recall"

    ].append(

        val_recall

    )


    history[

        "val_f1"

    ].append(

        val_f1

    )


    # --------------------------------------------------------
    # PRINT
    # --------------------------------------------------------

    print("\n")

    print(

        f"Train Loss     : "
        f"{train_loss:.4f}"

    )


    print(

        f"Train Accuracy : "
        f"{train_accuracy:.4f}"

    )


    print(

        f"Val Loss       : "
        f"{val_loss:.4f}"

    )


    print(

        f"Val Accuracy   : "
        f"{val_accuracy:.4f}"

    )


    print(

        f"Val Precision  : "
        f"{val_precision:.4f}"

    )


    print(

        f"Val Recall     : "
        f"{val_recall:.4f}"

    )


    print(

        f"Val F1         : "
        f"{val_f1:.4f}"

    )


    print(

        f"Learning Rate  : "
        f"{optimizer.param_groups[0]['lr']:.8f}"

    )


    # --------------------------------------------------------
    # GPU MEMORY
    # --------------------------------------------------------

    if DEVICE.type == "cuda":

        allocated = (

            torch.cuda.memory_allocated()

            /

            (1024 ** 3)

        )


        reserved = (

            torch.cuda.memory_reserved()

            /

            (1024 ** 3)

        )


        print(

            f"GPU Memory     : "
            f"{allocated:.2f} GB allocated"

        )


        print(

            f"GPU Reserved   : "
            f"{reserved:.2f} GB"

        )


    # --------------------------------------------------------
    # SAVE BEST
    # --------------------------------------------------------

    if val_f1 > best_f1:

        best_f1 = val_f1


        torch.save(

            {

                "model_state_dict":
                    model.state_dict(),

                "best_f1":
                    best_f1,

                "epoch":
                    epoch + 1,

                "class_names":
                    CLASS_NAMES,

                "alpha":
                    ALPHA,

                "image_size":
                    IMAGE_SIZE

            },

            "best_FERL_ViT_Maize.pth"

        )


        print(

            "\n>>> BEST MODEL SAVED"

        )


# ============================================================
# 36. LOAD BEST MODEL
# ============================================================

checkpoint = torch.load(

    "best_FERL_ViT_Maize.pth",

    map_location=DEVICE

)


model.load_state_dict(

    checkpoint[

        "model_state_dict"

    ]

)


print("\n")
print("=" * 70)
print("BEST MODEL LOADED")
print("=" * 70)


print(

    "Best epoch:",

    checkpoint["epoch"]

)


print(

    "Best validation F1:",

    checkpoint["best_f1"]

)


# ============================================================
# 37. FINAL TEST
# ============================================================

(

    test_loss,

    test_accuracy,

    test_precision,

    test_recall,

    test_f1,

    y_true,

    y_pred

) = evaluate(

    test_loader

)


print("\n")
print("=" * 70)
print("FINAL TEST RESULTS")
print("=" * 70)


print(

    f"Test Loss      : "
    f"{test_loss:.4f}"

)


print(

    f"Test Accuracy  : "
    f"{test_accuracy:.4f}"

)


print(

    f"Test Precision : "
    f"{test_precision:.4f}"

)


print(

    f"Test Recall    : "
    f"{test_recall:.4f}"

)


print(

    f"Test F1 Score  : "
    f"{test_f1:.4f}"

)


# ============================================================
# 38. CLASSIFICATION REPORT
# ============================================================

report = classification_report(

    y_true,

    y_pred,

    target_names=CLASS_NAMES,

    digits=4,

    zero_division=0

)


print("\n")
print("=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)


print(

    report

)


with open(

    "FERL_ViT_Maize_Classification_Report.txt",

    "w"

) as f:

    f.write(

        report

    )


# ============================================================
# 39. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_true,

    y_pred

)


print("\n")
print("=" * 70)
print("CONFUSION MATRIX")
print("=" * 70)


print(

    cm

)


plt.figure(

    figsize=(9, 7)

)


plt.imshow(

    cm,

    interpolation="nearest",

    cmap="Blues"

)


plt.title(

    "FERL-ViT Maize Confusion Matrix"

)


plt.colorbar()


ticks = np.arange(

    NUM_CLASSES

)


plt.xticks(

    ticks,

    CLASS_NAMES,

    rotation=30,

    ha="right"

)


plt.yticks(

    ticks,

    CLASS_NAMES

)


threshold = cm.max() / 2


for i in range(

    NUM_CLASSES

):

    for j in range(

        NUM_CLASSES

    ):

        plt.text(

            j,

            i,

            str(

                cm[i, j]

            ),

            ha="center",

            va="center",

            color=(

                "white"

                if cm[i, j] > threshold

                else "black"

            )

        )


plt.xlabel(

    "Predicted Class"

)


plt.ylabel(

    "Actual Class"

)


plt.tight_layout()


plt.savefig(

    "FERL_ViT_Maize_Confusion_Matrix.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 40. NORMALIZED CONFUSION MATRIX
# ============================================================

row_sums = cm.sum(

    axis=1,

    keepdims=True

)


row_sums[

    row_sums == 0

] = 1


cm_normalized = (

    cm.astype(

        np.float32

    )

    /

    row_sums

)


plt.figure(

    figsize=(9, 7)

)


plt.imshow(

    cm_normalized,

    interpolation="nearest",

    cmap="Blues"

)


plt.title(

    "FERL-ViT Maize Normalized Confusion Matrix"

)


plt.colorbar()


plt.xticks(

    ticks,

    CLASS_NAMES,

    rotation=30,

    ha="right"

)


plt.yticks(

    ticks,

    CLASS_NAMES

)


for i in range(

    NUM_CLASSES

):

    for j in range(

        NUM_CLASSES

    ):

        plt.text(

            j,

            i,

            f"{cm_normalized[i, j]:.2f}",

            ha="center",

            va="center"

        )


plt.xlabel(

    "Predicted Class"

)


plt.ylabel(

    "Actual Class"

)


plt.tight_layout()


plt.savefig(

    "FERL_ViT_Maize_Normalized_CM.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 41. ACCURACY CURVE
# ============================================================

plt.figure(

    figsize=(8, 6)

)


plt.plot(

    history["train_accuracy"],

    label="Training Accuracy",

    linewidth=2

)


plt.plot(

    history["val_accuracy"],

    label="Validation Accuracy",

    linewidth=2

)


plt.xlabel(

    "Epoch"

)


plt.ylabel(

    "Accuracy"

)


plt.title(

    "FERL-ViT Training and Validation Accuracy"

)


plt.legend()


plt.grid(

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    "FERL_ViT_Maize_Accuracy.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 42. LOSS CURVE
# ============================================================

plt.figure(

    figsize=(8, 6)

)


plt.plot(

    history["train_loss"],

    label="Training Loss",

    linewidth=2

)


plt.plot(

    history["val_loss"],

    label="Validation Loss",

    linewidth=2

)


plt.xlabel(

    "Epoch"

)


plt.ylabel(

    "Loss"

)


plt.title(

    "FERL-ViT Training and Validation Loss"

)


plt.legend()


plt.grid(

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    "FERL_ViT_Maize_Loss.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 43. F1 CURVE
# ============================================================

plt.figure(

    figsize=(8, 6)

)


plt.plot(

    history["val_f1"],

    label="Validation F1",

    linewidth=2

)


plt.xlabel(

    "Epoch"

)


plt.ylabel(

    "F1 Score"

)


plt.title(

    "FERL-ViT Validation F1 Score"

)


plt.legend()


plt.grid(

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    "FERL_ViT_Maize_F1.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 44. PRECISION AND RECALL
# ============================================================

plt.figure(

    figsize=(8, 6)

)


plt.plot(

    history["val_precision"],

    label="Validation Precision",

    linewidth=2

)


plt.plot(

    history["val_recall"],

    label="Validation Recall",

    linewidth=2

)


plt.xlabel(

    "Epoch"

)


plt.ylabel(

    "Score"

)


plt.title(

    "FERL-ViT Precision and Recall"

)


plt.legend()


plt.grid(

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    "FERL_ViT_Maize_Precision_Recall.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 45. SAVE FINAL RESULTS
# ============================================================

with open(

    "FERL_ViT_Maize_Final_Results.txt",

    "w"

) as f:

    f.write(

        "FERL-ViT Maize Classification\n"

    )

    f.write(

        "=============================\n\n"

    )

    f.write(

        f"Dataset: {ZIP_PATH}\n"

    )

    f.write(

        "Split: 70% Train / 15% Validation / 15% Test\n"

    )

    f.write(

        f"Epochs: {NUM_EPOCHS}\n"

    )

    f.write(

        f"Image Size: {IMAGE_SIZE}\n"

    )

    f.write(

        f"Batch Size: {BATCH_SIZE}\n"

    )

    f.write(

        f"Learning Rate: {LEARNING_RATE}\n"

    )

    f.write(

        f"Alpha: {ALPHA}\n\n"

    )

    f.write(

        f"Test Accuracy: "
        f"{test_accuracy:.6f}\n"

    )

    f.write(

        f"Test Precision: "
        f"{test_precision:.6f}\n"

    )

    f.write(

        f"Test Recall: "
        f"{test_recall:.6f}\n"

    )

    f.write(

        f"Test F1: "
        f"{test_f1:.6f}\n\n"

    )

    f.write(

        "Confusion Matrix:\n"

    )

    f.write(

        str(cm)

    )


# ============================================================
# 46. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("TRAINING COMPLETED")
print("=" * 70)


print(

    f"Dataset           : "
    f"{os.path.basename(ZIP_PATH)}"

)


print(

    f"Total images      : "
    f"{len(image_paths)}"

)


print(

    f"Training images   : "
    f"{len(train_paths)}"

)


print(

    f"Validation images : "
    f"{len(val_paths)}"

)


print(

    f"Testing images    : "
    f"{len(test_paths)}"

)


print(

    f"Epochs            : "
    f"{NUM_EPOCHS}"

)


print(

    f"Test Accuracy     : "
    f"{test_accuracy * 100:.2f}%"

)


print(

    f"Test Precision    : "
    f"{test_precision * 100:.2f}%"

)


print(

    f"Test Recall       : "
    f"{test_recall * 100:.2f}%"

)


print(

    f"Test F1           : "
    f"{test_f1 * 100:.2f}%"

)


if DEVICE.type == "cuda":

    print(

        f"GPU               : "
        f"{torch.cuda.get_device_name(0)}"

    )


print("\n")
print("OUTPUT FILES")
print("-" * 70)


print(

    "best_FERL_ViT_Maize.pth"

)


print(

    "FERL_ViT_Maize_Classification_Report.txt"

)


print(

    "FERL_ViT_Maize_Final_Results.txt"

)


print(

    "FERL_ViT_Maize_Confusion_Matrix.png"

)


print(

    "FERL_ViT_Maize_Normalized_CM.png"

)


print(

    "FERL_ViT_Maize_Accuracy.png"

)


print(

    "FERL_ViT_Maize_Loss.png"

)


print(

    "FERL_ViT_Maize_F1.png"

)


print(

    "FERL_ViT_Maize_Precision_Recall.png"

)


print("\n")
print("DONE.")
```
