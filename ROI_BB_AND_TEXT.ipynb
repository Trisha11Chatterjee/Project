```python
# ============================================================
# PLANTVILLAGE - ROI + BOUNDING BOX + PLANT NAME
# ============================================================
#
# 13 PlantVillage plant classes:
# Apple, Blueberry, Cherry, Corn, Grape, Peach, Pepper,
# Potato, Raspberry, Soybean, Squash, Strawberry, Tomato
#
# Pipeline:
#
# PlantVillage
#      |
#      v
# Find plant/disease folders
#      |
#      v
# Get plant name from folder
#      |
#      v
# Read image
#      |
#      v
# OpenCV HSV ROI segmentation
#      |
#      v
# Morphological processing
#      |
#      v
# Contour detection
#      |
#      v
# Leaf bounding box
#      |
#      v
# Draw bounding box
#      |
#      v
# Write plant name
#      |
#      v
# Save annotated image
#
# IMPORTANT:
# This is NOT CNN classification.
# The plant label is inherited from the PlantVillage folder.
# ============================================================


# ============================================================
# CELL 1: INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install -q kagglehub opencv-python matplotlib pandas tqdm


# ============================================================
# CELL 2: IMPORT LIBRARIES
# ============================================================

import os
import cv2
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tqdm.auto import tqdm
from collections import Counter


# ============================================================
# CELL 3: DOWNLOAD PLANTVILLAGE USING KAGGLEHUB
# ============================================================

import kagglehub

print("Downloading PlantVillage dataset...")
print("Please wait...")

dataset_path = kagglehub.dataset_download(
    "mohitsingh1804/plantvillage"
)

print("\nDataset downloaded successfully.")
print("Dataset path:")
print(dataset_path)


# ============================================================
# CELL 4: DEFINE THE 13 PLANT CLASSES
# ============================================================

PLANT_NAMES = [
    "Apple",
    "Blueberry",
    "Cherry",
    "Corn",
    "Grape",
    "Peach",
    "Pepper",
    "Potato",
    "Raspberry",
    "Soybean",
    "Squash",
    "Strawberry",
    "Tomato"
]

print("\n13 Target Plant Classes")
print("=" * 40)

for i, plant in enumerate(PLANT_NAMES, start=1):
    print(f"{i:2d}. {plant}")


# ============================================================
# CELL 5: FIND ALL PLANTVILLAGE FOLDERS
# ============================================================
#
# PlantVillage commonly contains folders such as:
#
# Apple___Apple_scab
# Apple___Black_rot
# Apple___healthy
#
# Potato___Early_blight
# Potato___Late_blight
# Potato___healthy
#
# We extract the plant name from the text before "___".
# ============================================================

plant_folders = {
    plant: []
    for plant in PLANT_NAMES
}

print("\nSearching dataset folders...")
print("=" * 60)

for root, dirs, files in os.walk(dataset_path):

    for directory in dirs:

        # Extract plant name
        plant_name = directory.split("___")[0]

        if plant_name in PLANT_NAMES:

            full_folder_path = os.path.join(
                root,
                directory
            )

            plant_folders[plant_name].append(
                full_folder_path
            )


# ============================================================
# CELL 6: DISPLAY FOUND FOLDERS
# ============================================================

print("\nFolders found for each plant:")
print("=" * 60)

for plant in PLANT_NAMES:

    folders = plant_folders[plant]

    print(
        f"\n{plant}: {len(folders)} folder(s)"
    )

    for folder in folders:
        print("   ", folder)


# ============================================================
# CELL 7: CHECK ALL 13 CLASSES
# ============================================================

missing_classes = [
    plant
    for plant in PLANT_NAMES
    if len(plant_folders[plant]) == 0
]

print("\n" + "=" * 60)

if len(missing_classes) == 0:

    print("SUCCESS")
    print("All 13 PlantVillage plant classes were found.")

else:

    print("WARNING")
    print("The following classes were not found:")

    for plant in missing_classes:
        print(" -", plant)


# ============================================================
# CELL 8: COLLECT ALL IMAGE PATHS
# ============================================================

VALID_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
)

image_records = []

print("\nCollecting image paths...")

for plant_name in PLANT_NAMES:

    for folder in plant_folders[plant_name]:

        try:
            files = os.listdir(folder)
        except Exception:
            continue

        for filename in files:

            if filename.lower().endswith(
                VALID_EXTENSIONS
            ):

                image_path = os.path.join(
                    folder,
                    filename
                )

                image_records.append({
                    "image_path": image_path,
                    "plant_name": plant_name,
                    "source_folder": os.path.basename(folder)
                })


# ============================================================
# CELL 9: DISPLAY TOTAL IMAGE COUNT
# ============================================================

print("\nTotal images found:")
print(len(image_records))


# ============================================================
# CELL 10: IMAGE COUNT PER PLANT
# ============================================================

class_counts = Counter(
    record["plant_name"]
    for record in image_records
)

print("\nImage count for each plant")
print("=" * 50)

for plant in PLANT_NAMES:

    count = class_counts.get(
        plant,
        0
    )

    print(
        f"{plant:12s} : {count:6d}"
    )


# ============================================================
# CELL 11: CREATE OUTPUT DIRECTORIES
# ============================================================

OUTPUT_DIR = "/content/PlantVillage_BoundingBox"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

for plant in PLANT_NAMES:

    os.makedirs(
        os.path.join(
            OUTPUT_DIR,
            plant
        ),
        exist_ok=True
    )

print("\nOutput directory:")
print(OUTPUT_DIR)


# ============================================================
# CELL 12: OPENCV ROI + BOUNDING BOX FUNCTION
# ============================================================

def detect_leaf_and_draw_bbox(
    image,
    plant_name,
    min_area_ratio=0.01,
    margin_ratio=0.02
):

    # --------------------------------------------------------
    # Original image
    # --------------------------------------------------------

    original = image.copy()

    height, width = image.shape[:2]

    image_area = height * width

    # --------------------------------------------------------
    # Convert BGR -> HSV
    # --------------------------------------------------------

    hsv = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2HSV
    )

    # --------------------------------------------------------
    # Extract saturation channel
    # --------------------------------------------------------

    saturation = hsv[:, :, 1]

    # --------------------------------------------------------
    # Threshold saturation
    # --------------------------------------------------------

    _, mask = cv2.threshold(
        saturation,
        30,
        255,
        cv2.THRESH_BINARY
    )

    # --------------------------------------------------------
    # Morphological processing
    # --------------------------------------------------------

    kernel = np.ones(
        (5, 5),
        np.uint8
    )

    # Remove small noise
    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        kernel,
        iterations=1
    )

    # Fill small holes
    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        kernel,
        iterations=2
    )

    # --------------------------------------------------------
    # Find contours
    # --------------------------------------------------------

    contours, _ = cv2.findContours(
        mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    # --------------------------------------------------------
    # If no contour is found
    # --------------------------------------------------------

    if len(contours) == 0:

        output = original.copy()

        label = plant_name

        cv2.putText(
            output,
            label,
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1.0,
            (0, 255, 0),
            2,
            cv2.LINE_AA
        )

        return output, None

    # --------------------------------------------------------
    # Minimum contour area
    # --------------------------------------------------------

    min_area = (
        image_area *
        min_area_ratio
    )

    valid_contours = []

    for contour in contours:

        area = cv2.contourArea(
            contour
        )

        if area >= min_area:

            valid_contours.append(
                contour
            )

    # --------------------------------------------------------
    # Select largest valid contour
    # --------------------------------------------------------

    if len(valid_contours) > 0:

        selected_contour = max(
            valid_contours,
            key=cv2.contourArea
        )

    else:

        selected_contour = max(
            contours,
            key=cv2.contourArea
        )

    # --------------------------------------------------------
    # Bounding rectangle
    # --------------------------------------------------------

    x, y, w, h = cv2.boundingRect(
        selected_contour
    )

    # --------------------------------------------------------
    # Add margin around leaf
    # --------------------------------------------------------

    margin_x = int(
        width *
        margin_ratio
    )

    margin_y = int(
        height *
        margin_ratio
    )

    x1 = max(
        0,
        x - margin_x
    )

    y1 = max(
        0,
        y - margin_y
    )

    x2 = min(
        width,
        x + w + margin_x
    )

    y2 = min(
        height,
        y + h + margin_y
    )

    # --------------------------------------------------------
    # Draw bounding box
    # --------------------------------------------------------

    output = original.copy()

    cv2.rectangle(
        output,
        (x1, y1),
        (x2, y2),
        (0, 255, 0),
        3
    )

    # --------------------------------------------------------
    # Prepare label
    # --------------------------------------------------------

    label = plant_name

    font = cv2.FONT_HERSHEY_SIMPLEX

    # Adjust font according to image size
    font_scale = max(
        0.6,
        width / 1200
    )

    thickness = 2

    (
        text_width,
        text_height
    ), baseline = cv2.getTextSize(
        label,
        font,
        font_scale,
        thickness
    )

    # --------------------------------------------------------
    # Label position
    # --------------------------------------------------------

    label_x = x1

    label_y = max(
        text_height + baseline + 5,
        y1
    )

    # --------------------------------------------------------
    # Label background rectangle
    # --------------------------------------------------------

    cv2.rectangle(
        output,

        (
            label_x,
            label_y -
            text_height -
            baseline -
            5
        ),

        (
            label_x +
            text_width +
            10,
            label_y + 5
        ),

        (0, 255, 0),

        -1
    )

    # --------------------------------------------------------
    # Write plant name
    # --------------------------------------------------------

    cv2.putText(
        output,
        label,
        (
            label_x + 5,
            label_y
        ),
        font,
        font_scale,
        (0, 0, 0),
        thickness,
        cv2.LINE_AA
    )

    # --------------------------------------------------------
    # Store bounding-box coordinates
    # --------------------------------------------------------

    bbox = {
        "x1": int(x1),
        "y1": int(y1),
        "x2": int(x2),
        "y2": int(y2),
        "bbox_width": int(x2 - x1),
        "bbox_height": int(y2 - y1)
    }

    return output, bbox


# ============================================================
# CELL 13: TEST ROI ON ONE IMAGE FIRST
# ============================================================

if len(image_records) > 0:

    test_record = random.choice(
        image_records
    )

    test_image = cv2.imread(
        test_record["image_path"]
    )

    test_output, test_bbox = (
        detect_leaf_and_draw_bbox(
            test_image,
            test_record["plant_name"]
        )
    )

    # Convert BGR -> RGB
    test_rgb = cv2.cvtColor(
        test_output,
        cv2.COLOR_BGR2RGB
    )

    plt.figure(
        figsize=(8, 8)
    )

    plt.imshow(
        test_rgb
    )

    plt.title(
        f"Plant: {test_record['plant_name']}"
    )

    plt.axis("off")

    plt.show()

    print("Test image:")
    print(test_record["image_path"])

    print("\nBounding box:")
    print(test_bbox)


# ============================================================
# CELL 14: PROCESS ALL IMAGES
# ============================================================

results = []

print("\nStarting complete image processing...")
print(
    f"Total images to process: {len(image_records)}"
)

for record in tqdm(
    image_records,
    desc="Creating bounding boxes"
):

    image_path = record["image_path"]

    plant_name = record["plant_name"]

    source_folder = record[
        "source_folder"
    ]

    # --------------------------------------------------------
    # Read image
    # --------------------------------------------------------

    image = cv2.imread(
        image_path
    )

    if image is None:

        results.append({
            "original_path": image_path,
            "output_path": "",
            "plant_name": plant_name,
            "source_folder": source_folder,
            "x1": None,
            "y1": None,
            "x2": None,
            "y2": None,
            "bbox_width": None,
            "bbox_height": None,
            "status": "READ_ERROR"
        })

        continue

    # --------------------------------------------------------
    # Detect leaf and draw bounding box
    # --------------------------------------------------------

    annotated_image, bbox = (
        detect_leaf_and_draw_bbox(
            image=image,
            plant_name=plant_name
        )
    )

    # --------------------------------------------------------
    # Original filename
    # --------------------------------------------------------

    base_name = os.path.splitext(
        os.path.basename(
            image_path
        )
    )[0]

    # --------------------------------------------------------
    # Make unique filename
    # --------------------------------------------------------

    safe_source_folder = (
        source_folder
        .replace(
            "___",
            "_"
        )
        .replace(
            " ",
            "_"
        )
    )

    output_filename = (
        f"{safe_source_folder}"
        f"_{base_name}"
        f"_bbox.jpg"
    )

    # --------------------------------------------------------
    # Output directory
    # --------------------------------------------------------

    plant_output_dir = os.path.join(
        OUTPUT_DIR,
        plant_name
    )

    output_path = os.path.join(
        plant_output_dir,
        output_filename
    )

    # --------------------------------------------------------
    # Save image
    # --------------------------------------------------------

    saved = cv2.imwrite(
        output_path,
        annotated_image
    )

    # --------------------------------------------------------
    # Store metadata
    # --------------------------------------------------------

    result = {
        "original_path": image_path,
        "output_path": output_path,
        "plant_name": plant_name,
        "source_folder": source_folder
    }

    if bbox is not None:

        result.update(bbox)

        result["status"] = (
            "BOUNDING_BOX_DETECTED"
        )

    else:

        result.update({
            "x1": None,
            "y1": None,
            "x2": None,
            "y2": None,
            "bbox_width": None,
            "bbox_height": None
        })

        result["status"] = (
            "NO_CONTOUR"
        )

    results.append(
        result
    )


# ============================================================
# CELL 15: PROCESSING SUMMARY
# ============================================================

results_df = pd.DataFrame(
    results
)

print("\n")
print("=" * 70)
print("PROCESSING COMPLETED")
print("=" * 70)

print(
    "Total records:",
    len(results_df)
)

print(
    "Successfully saved:",
    sum(
        results_df["output_path"]
        != ""
    )
)

print(
    "Bounding boxes detected:",
    sum(
        results_df["status"]
        == "BOUNDING_BOX_DETECTED"
    )
)

print(
    "No contour:",
    sum(
        results_df["status"]
        == "NO_CONTOUR"
    )
)

print(
    "Read errors:",
    sum(
        results_df["status"]
        == "READ_ERROR"
    )
)


# ============================================================
# CELL 16: SAVE BOUNDING BOX CSV
# ============================================================

CSV_PATH = os.path.join(
    OUTPUT_DIR,
    "PlantVillage_BoundingBoxes.csv"
)

results_df.to_csv(
    CSV_PATH,
    index=False
)

print("\nCSV saved:")
print(CSV_PATH)


# ============================================================
# CELL 17: DISPLAY CSV INFORMATION
# ============================================================

print("\nFirst 10 records:")

display(
    results_df.head(10)
)


# ============================================================
# CELL 18: BOUNDING BOX DETECTION RATE
# ============================================================

total_images = len(
    results_df
)

detected_images = sum(
    results_df["status"]
    == "BOUNDING_BOX_DETECTED"
)

if total_images > 0:

    detection_rate = (
        detected_images /
        total_images *
        100
    )

else:

    detection_rate = 0


print(
    f"\nBounding-box detection rate: "
    f"{detection_rate:.2f}%"
)


# ============================================================
# CELL 19: BOUNDING BOX RESULTS PER CLASS
# ============================================================

class_summary = []

for plant in PLANT_NAMES:

    class_data = results_df[
        results_df["plant_name"]
        == plant
    ]

    total = len(
        class_data
    )

    detected = sum(
        class_data["status"]
        == "BOUNDING_BOX_DETECTED"
    )

    if total > 0:

        rate = (
            detected /
            total *
            100
        )

    else:

        rate = 0

    class_summary.append({
        "Plant": plant,
        "Total Images": total,
        "Bounding Boxes": detected,
        "Detection Rate (%)": round(
            rate,
            2
        )
    })


class_summary_df = pd.DataFrame(
    class_summary
)

print("\nBounding-box results per plant:")

display(
    class_summary_df
)


# ============================================================
# CELL 20: SHOW RANDOM ANNOTATED IMAGES
# ============================================================

def show_random_results(
    dataframe,
    n=12
):

    valid_data = dataframe[
        dataframe["output_path"]
        != ""
    ]

    if len(valid_data) == 0:

        print(
            "No output images available."
        )

        return

    n = min(
        n,
        len(valid_data)
    )

    sample = valid_data.sample(
        n=n,
        random_state=42
    )

    rows = 3

    cols = 4

    plt.figure(
        figsize=(20, 15)
    )

    for i, (_, row) in enumerate(
        sample.iterrows()
    ):

        image = cv2.imread(
            row["output_path"]
        )

        if image is None:
            continue

        image = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB
        )

        ax = plt.subplot(
            rows,
            cols,
            i + 1
        )

        ax.imshow(
            image
        )

        ax.set_title(
            row["plant_name"],
            fontsize=14
        )

        ax.axis("off")

    plt.tight_layout()

    plt.show()


show_random_results(
    results_df,
    n=12
)


# ============================================================
# CELL 21: SHOW ONE EXAMPLE FROM EACH OF THE 13 CLASSES
# ============================================================

print(
    "\nShowing one bounding-box result "
    "from each plant class..."
)

plt.figure(
    figsize=(20, 25)
)

plot_index = 1

for plant in PLANT_NAMES:

    plant_data = results_df[
        (
            results_df["plant_name"]
            == plant
        )
        &
        (
            results_df["output_path"]
            != ""
        )
    ]

    if len(plant_data) == 0:
        continue

    row = plant_data.iloc[0]

    image = cv2.imread(
        row["output_path"]
    )

    if image is None:
        continue

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    ax = plt.subplot(
        4,
        4,
        plot_index
    )

    ax.imshow(
        image
    )

    ax.set_title(
        plant,
        fontsize=14
    )

    ax.axis("off")

    plot_index += 1


plt.tight_layout()

plt.show()


# ============================================================
# CELL 22: PRINT OUTPUT FOLDER STRUCTURE
# ============================================================

print("\nOutput folder structure:")
print("=" * 60)

for plant in PLANT_NAMES:

    folder = os.path.join(
        OUTPUT_DIR,
        plant
    )

    if os.path.exists(folder):

        count = len([
            f
            for f in os.listdir(folder)
            if f.lower().endswith(
                (".jpg", ".jpeg", ".png")
            )
        ])

        print(
            f"{plant:12s} -> "
            f"{count} annotated images"
        )


# ============================================================
# CELL 23: CREATE ZIP FILE
# ============================================================

import shutil

ZIP_BASE = "/content/PlantVillage_BoundingBox"

ZIP_PATH = shutil.make_archive(
    ZIP_BASE,
    "zip",
    OUTPUT_DIR
)

print("\nZIP file created:")
print(ZIP_PATH)


# ============================================================
# CELL 24: FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print("Dataset:")
print("PlantVillage")

print("\nPlant classes:")
for i, plant in enumerate(
    PLANT_NAMES,
    start=1
):
    print(
        f"{i:2d}. {plant}"
    )

print("\nProcessing:")
print("OpenCV ROI segmentation")
print("Contour detection")
print("Leaf bounding box")
print("Plant-name annotation")

print("\nCNN:")
print("NOT USED")

print("\nOutput directory:")
print(OUTPUT_DIR)

print("\nCSV:")
print(CSV_PATH)

print("\nZIP:")
print(ZIP_PATH)

print("\nProcessing completed successfully.")
```
